```python
"""
Код HW3_MV.ipynb (далее - код) представляет собой автоматизированный конвейер обработки ежедневных данных для анализа финансовых инструментов,
таких как акции и криптовалюты.
Основная цель конвейера — сбор, обработка, нормализация и анализ ежедневных данных с последующей генерацией признаков
для построения моделей машинного обучения.
Конвейер также включает визуализацию данных, оценку важности признаков и автоматическое обновление данных с заданным интервалом.
Код использует библиотеки для работы с данными (например, pandas, numpy), визуализации (matplotlib, seaborn), машинного обучения (scikit-learn, shap),
а также для автоматизации задач (schedule).
Процесс обработки данных управляется через графический интерфейс, который отображает статус выполнения задач и позволяет пользователю контролировать процесс.

Конвейер состоит из нескольких этапов:
Загрузка данных — сбор ежедневных данных о ценах на акции и криптовалюты с использованием API Yahoo Finance.
Обработка данных — очистка данных, удаление выбросов, заполнение пропусков и устранение дубликатов.
Нормализация данных — применение различных методов нормализации для подготовки данных к анализу.
Генерация признаков — создание новых признаков на основе технических индикаторов и статистических методов.
Оценка важности признаков — анализ значимости признаков для построения моделей машинного обучения.
Автоматическое обновление — периодическая проверка наличия новых данных и их обработка.

Код сохраняет загруженные данные в папку data/raw, а обработанные данные - в папку data/processed.

Конвейер может работать как в ручном, так и в автоматическом режиме, при введении пользователем необходимых параметров работы.

Copyright (c) 2025 Matvei Vasetsov (Матвей Васецов). All rights reserved.
Этот код является интеллектуальной собственностью и защищен авторским правом. Любое использование, копирование или распространение без разрешения автора запрещено.

"""
import time
import schedule
from datetime import datetime, timedelta
import sys
import pandas as pd
import numpy as np
import yfinance as yf
import ta
import os
import datetime
import matplotlib.pyplot as plt
import matplotlib.dates as mdates
from matplotlib.ticker import MaxNLocator 
import seaborn as sns
import featuretools as ft
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from scipy import stats
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.inspection import permutation_importance
import shap
import warnings
from tsfresh import extract_features
from tsfresh.utilities.dataframe_functions import impute
import tkinter as tk
from tkinter import ttk, messagebox
from matplotlib.figure import Figure
from matplotlib.backends.backend_tkagg import FigureCanvasTkAgg
import shutil

warnings.filterwarnings('ignore')

class DataPipeline:
    def __init__(self, stocks=['MSTR', 'BRPHF'], cryptos=['BTC-USD', 'SOL-USD']):
        """
        Инициализация конвейера данных. 
        Создаются необходимые директории для хранения данных, инициализируется таблица статусов для отслеживания выполнения задач,
        а также создается графический интерфейс для отображения статуса и управления процессом.
        """
        self.current_date = datetime.datetime.now()
        self.interval = '1d'
        self.start_date = '2020-01-01'
        self.end_date = self.current_date.strftime('%Y-%m-%d')
        
        self.stocks = stocks
        self.cryptos = cryptos
        self.instruments = stocks + cryptos
        
        # Удаление старой и создание директории для хранения данных
        if not os.path.exists('data'):
            os.makedirs('data/raw', exist_ok=True)
            os.makedirs('data/processed', exist_ok=True)
        
        # Расширенный статус для дашборда
        self.status = pd.DataFrame(
            index=self.instruments,
            columns=[
                'Last Update', 'Downloaded', 'Processed', 'Normalized', 
                'Features Generated', 'Feature Importance',
                'Check Status', 'Old Data Count', 'New Data Count', 
                'Price Change %'
            ]
        )
        self.status.fillna('Pending', inplace=True)
        
        # Инициализация окна Tkinter
        self.root = tk.Tk()
        self.root.title("Enhanced Data Pipeline Dashboard")
        self.root.geometry("1200x300")  # Уменьшена высота окна, так как убрана область графика
        
        # Создание фреймов
        self.create_dashboard_frames()
        
        # Флаги для контроля выполнения
        self.is_running = True
        self.waiting_for_data = False
        
        # Параметры автообновления
        self.check_interval = None
        self.max_checks = None
        self.current_check = 0
        
        self.root.protocol("WM_DELETE_WINDOW", self.on_close)
        
    def create_dashboard_frames(self):
        """
        Создание интерфейса дашборда, который включает таблицу для отображения статусов выполнения задач и кнопки управления.
        Интерфейс позволяет пользователю видеть текущее состояние конвейера и взаимодействовать с ним.
        """
        # Верхняя панель для таблицы
        self.table_frame = ttk.Frame(self.root)
        self.table_frame.pack(pady=10, padx=10, fill='both', expand=True)
        
        # Создание таблицы
        self.tree = ttk.Treeview(
            self.table_frame, 
            columns=['Instrument'] + list(self.status.columns), 
            show='headings',
            height=10
        )
        
        # Настройка заголовков и столбцов
        self.tree.heading('Instrument', text='Symbol')
        for col in self.status.columns:
            self.tree.heading(col, text=col)
            self.tree.column(col, width=90)
        
        # Добавление скроллбаров
        y_scroll = ttk.Scrollbar(self.table_frame, orient='vertical', command=self.tree.yview)
        x_scroll = ttk.Scrollbar(self.table_frame, orient='horizontal', command=self.tree.xview)
        self.tree.configure(yscrollcommand=y_scroll.set, xscrollcommand=x_scroll.set)
        
        # Размещение элементов
        y_scroll.pack(side='right', fill='y')
        x_scroll.pack(side='bottom', fill='x')
        self.tree.pack(fill='both', expand=True)
        
        # Фрейм для кнопок управления
        self.control_frame = ttk.Frame(self.root)
        self.control_frame.pack(pady=5, padx=10, fill='x')
        
        # Кнопка OK для подтверждения
        self.ok_button = ttk.Button(
            self.control_frame, 
            text="OK", 
            command=self.handle_ok_button
        )
        self.ok_button.pack(side='right', padx=5)
        self.ok_button.pack_forget()  # Скрываем кнопку initially

    def handle_ok_button(self):
        """
        Обработчик нажатия кнопки "OK" в интерфейсе. Используется для подтверждения действий пользователя,
        например, при ожидании следующей проверки данных.
        """
        if self.waiting_for_data:
            self.ok_button.pack_forget()
            messagebox.showinfo(
                "Ожидание", 
                f"Ожидание {self.check_interval} мин. перед следующей проверкой..."
            )
            self.root.update()
    
    def update_dashboard(self, old_data=None, new_data=None, instrument=None):
        """
        Обновление интерфейса дашборда, включая таблицу статусов.
        Метод обеспечивает актуальное отображение состояния конвейера в реальном времени.
        """
        if not self.is_running:
            return
            
        try:
            # Очистка текущих данных в таблице
            for item in self.tree.get_children():
                self.tree.delete(item)
            
            # Добавление новых данных
            for instrument, row in self.status.iterrows():
                self.tree.insert('', 'end', values=[instrument] + list(row))
            
            # Обновление окна
            if self.is_running:
                self.root.update()
                
        except tk.TclError:
            pass
  
    def download_data(self, mode='manual'):
        """
        Загрузка данных о ценах на акции и криптовалюты с использованием API Yahoo Finance.
        Метод также проверяет наличие новых данных и обновляет статусы в таблице.
        """
        has_new_data = False
    
        for instrument in self.instruments:
            try:
                self.status.loc[instrument, 'Check Status'] = 'Checking'
                self.status.loc[instrument, 'Downloaded'] = 'In Progress'
                self.update_dashboard()
    
                # Загрузка новых данных
                new_data = yf.download(
                    instrument,
                    start=self.start_date,
                    end=self.end_date,
                    interval=self.interval
                )
                print(f"Загружены данные для {instrument}")
                
                if not new_data.empty:
                    new_data = new_data.reset_index()
                    new_data['Date'] = pd.to_datetime(new_data['Date'])
    
                    raw_file = f'data/raw/{instrument}_data.csv'
                    if os.path.exists(raw_file):
                        # Загрузка существующих данных
                        old_data = pd.read_csv(raw_file)
                        old_data['Date'] = pd.to_datetime(old_data['Date'])
    
                        # Определение новых данных
                        latest_date = old_data['Date'].max()
                        new_records = new_data[new_data['Date'] > latest_date]
    
                        # Обновление Old Data Count независимо от наличия новых данных
                        self.status.loc[instrument, 'Old Data Count'] = len(old_data)
    
                        if not new_records.empty:
                            # Обновление статуса
                            self.status.loc[instrument, 'New Data Count'] = len(new_records)
    
                            # Расчет изменения цены
                            old_price = old_data['Close'].iloc[-1]
                            new_price = new_records['Close'].iloc[-1]
                            price_change = ((new_price - old_price) / old_price) * 100
                            self.status.loc[instrument, 'Price Change %'] = f"{price_change:.2f}%"
    
                            # Объединение данных
                            data = pd.concat([old_data, new_records], ignore_index=True)
                            data = data.sort_values('Date')
                            has_new_data = True
    
                            # Обновление графика
                            self.update_dashboard(old_data, new_records, instrument)
                        else:
                            data = old_data
                            self.status.loc[instrument, 'Check Status'] = 'No New Data'
                            self.status.loc[instrument, 'New Data Count'] = 0
                    else:
                        data = new_data
                        has_new_data = True
                        self.status.loc[instrument, 'Check Status'] = 'Initial Data'
                        self.status.loc[instrument, 'New Data Count'] = len(data)
                        self.status.loc[instrument, 'Old Data Count'] = 0  # Нет старых данных
    
                    # Сохранение данных
                    data['Date'] = data['Date'].dt.strftime('%Y-%m-%d')
                    data.to_csv(raw_file, index=False)
    
                    self.status.loc[instrument, 'Downloaded'] = 'Complete'
                    self.status.loc[instrument, 'Last Update'] = datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S')
                else:
                    self.status.loc[instrument, 'Downloaded'] = 'No Data'
                    self.status.loc[instrument, 'Check Status'] = 'Failed'
                    self.status.loc[instrument, 'Old Data Count'] = 0  # Нет данных вообще
    
                self.update_dashboard()
    
            except Exception as e:
                self.status.loc[instrument, 'Downloaded'] = f'Error: {str(e)}'
                self.status.loc[instrument, 'Check Status'] = 'Error'
                self.update_dashboard()
    
        return has_new_data
    
    def show_correlation_matrix(self):
        """
        Построение и отображение матрицы корреляций между различными финансовыми инструментами.
        Метод помогает выявить взаимосвязи между активами.
        """
        print(f"\n\033[1;34mПостроение и отображение матрицы корреляций на основе загруженных данных \033[0m")
        try:
            # Загрузка всех загруженных данных
            all_data = []
            for instrument in self.instruments:
                raw_file = f'data/raw/{instrument}_data.csv'
                if os.path.exists(raw_file):
                    data = pd.read_csv(raw_file)
                    # Явное преобразование в datetime
                    data['Date'] = pd.to_datetime(data['Date'])
                    # Преобразование столбца Close в числовой формат
                    data['Close'] = pd.to_numeric(data['Close'], errors='coerce')
                    
                    # Удаление дубликатов по дате
                    data = data.drop_duplicates(subset=['Date'], keep='last')
                    
                    # Добавление данных в список
                    all_data.append(data.set_index('Date')['Close'].rename(instrument))
            
            if len(all_data) > 1:
                # Объединение данных по дате
                combined_data = pd.concat(all_data, axis=1, join='outer')
                
                # Заполнение пропущенных значений методом forward fill, затем backward fill
                combined_data = combined_data.fillna(method='ffill').fillna(method='bfill')
                
                # Расчет корреляционной матрицы
                correlation_matrix = combined_data.corr()
                
                # Визуализация корреляционной матрицы
                plt.figure(figsize=(10, 8))
                sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
                print("=" * 150)
                plt.title('Корреляционная матрица для загруженных данных ')
                plt.tight_layout()
                plt.show()
                
                # Сохранение корреляционной матрицы
                correlation_matrix.to_csv('data/correlation_matrix.csv')
                print(f"Корреляционная матрица сохранена в файл data/correlation_matrix.csv")
                print("=" * 150)
            else:
                print("Недостаточно данных для построения корреляционной матрицы.")
        except Exception as e:
            print(f"Ошибка при построении корреляционной матрицы: {e}")
    
    def process_data(self):
        """
        Обработка загруженных данных: очистка, удаление выбросов, заполнение пропусков и устранение дубликатов.
        Метод подготавливает данные для дальнейшего анализа.
        """
        print(f"\n\033[1;34mОбработка данных: очистка, заполнение пропусков, устранение дубликатов \033[0m")
        
        for instrument in self.instruments:
            try:
                raw_file = f'data/raw/{instrument}_data.csv'
                if not os.path.exists(raw_file):
                    print(f"Файл {raw_file} не найден. Пропуск обработки.")
                    continue
                
                data = pd.read_csv(raw_file)
                data['Date'] = pd.to_datetime(data['Date'])
                
                # Преобразование числовых столбцов
                numeric_columns = ['Close']
                for col in numeric_columns:
                    data[col] = pd.to_numeric(data[col], errors='coerce')
                
                # Обработка выбросов методом межквартильного размаха
                for column in numeric_columns:
                    if column in data.columns:
                        Q1 = data[column].quantile(0.25)
                        Q3 = data[column].quantile(0.75)
                        IQR = Q3 - Q1
                        lower_bound = Q1 - 1.5 * IQR
                        upper_bound = Q3 + 1.5 * IQR
                        data = data[(data[column] >= lower_bound) & (data[column] <= upper_bound)]
                print("Обработка выбросов методом межквартильного размаха завершена.")
                
                # Заполнение пропусков
                data = data.fillna(method='ffill').fillna(method='bfill')
                print("Заполнение пропусков завершено.")
                
                # Удаление дубликатов
                data = data.drop_duplicates()
                print("Удаление дубликатов завершено.")
                
                # Сохранение обработанных данных
                processed_file = f'data/processed/{instrument}_processed.csv'  
                data.to_csv(processed_file, index=False)
                print(f"Обработанные данные сохранены для {instrument} в файл {processed_file}")
                
                # Обновление статуса
                self.status.loc[instrument, 'Processed'] = 'Complete'
                self.update_dashboard()

            except Exception as e:
                print(f"Ошибка при обработке данных для {instrument}: {e}")
                self.status.loc[instrument, 'Processed'] = f'Error: {str(e)}'
                self.update_dashboard()
    
    def normalize_data(self):
        """
        Нормализация данных с использованием различных методов: Z-Score, Min-Max, Box-Cox, Log.
        Метод выбирает лучший метод нормализации на основе оценки качества.
        """
        print(f"\n\033[1;34mНормализация данных различными методами и выбор лучшего метода нормализации \033[0m")
        
        normalization_methods = {
            'Z-Score': StandardScaler(),
            'Min-Max': MinMaxScaler(),
            'Box-Cox': None,
            'Log': None
        }
        results = []
        self.best_normalization_methods = {}  # Словарь для хранения лучших методов нормализации
        for instrument in self.instruments:
            try:
                processed_file = f'data/processed/{instrument}_processed.csv'
                if not os.path.exists(processed_file):
                    print(f"Файл {processed_file} не найден. Пропуск нормализации.")
                    continue
                data = pd.read_csv(processed_file)
                data['Date'] = pd.to_datetime(data['Date'])
                X = data['Close'].values.reshape(-1, 1)
                best_method = None
                best_r2 = -np.inf
                for method_name, method in normalization_methods.items():
                    try:
                        if method_name == 'Box-Cox':
                            # Добавление малого значения для положительности
                            X_norm, lambda_param = stats.boxcox(X.flatten() + 1e-10)
                            X_norm = X_norm.reshape(-1, 1)
                        elif method_name == 'Log':
                            X_norm = np.log(X + 1e-10)
                        else:
                            X_norm = method.fit_transform(X)
                        
                        # Разделение данных
                        X_train, X_test = train_test_split(X_norm, test_size=0.2, shuffle=False)
                        
                        # Обучение простой модели для оценки качества нормализации
                        model = RandomForestRegressor(random_state=42)
                        model.fit(X_train[:-1], X_train[1:].ravel())
                        
                        # Прогнозирование
                        y_pred = model.predict(X_test[:-1])
                        
                        # Оценка качества
                        mae = mean_absolute_error(X_test[1:], y_pred)
                        mse = mean_squared_error(X_test[1:], y_pred)
                        r2 = r2_score(X_test[1:], y_pred)
                        results.append({
                            'Инструмент': instrument,
                            'Метод': method_name,
                            'MAE': mae,
                            'MSE': mse,
                            'R-square': r2
                        })
                        # Определение лучшего метода нормализации
                        if r2 > best_r2:
                            best_r2 = r2
                            best_method = method_name
                        
                        # Сохранение нормализованных данных
                        norm_df = pd.DataFrame(X_norm, columns=['Close_normalized'])
                        norm_df['Date'] = data['Date']  # Сохраняем дату
                        norm_df.to_csv(f'data/processed/{instrument}_{method_name}_normalized.csv', index=False)
                        print(f"Нормализованные данные методом {method_name} сохранены для {instrument} в файл data/processed/{instrument}_{method_name}_normalized.csv")
                        print("=" * 150)

                    except Exception as norm_error:
                        print(f"Ошибка при нормализации {method_name} для {instrument}: {norm_error}")
                
                # Сохранение лучшего метода нормализации для инструмента
                self.best_normalization_methods[instrument] = best_method
                print(f"Лучший метод нормализации для {instrument}: {best_method} (R-square: {best_r2:.4f})")
                # Обновление статуса
                self.status.loc[instrument, 'Normalized'] = 'Complete'
                self.update_dashboard()
                # Визуализация нормализованных данных
                normalized_file = f'data/processed/{instrument}_{best_method}_normalized.csv'
                if os.path.exists(normalized_file):
                    normalized_data = pd.read_csv(normalized_file)
                    normalized_data['Date'] = pd.to_datetime(normalized_data['Date'])
                    plt.figure(figsize=(18, 8))  # Увеличен размер по вертикали
                    
                    # График исходных данных
                    plt.subplot(1, 3, 1)
                    raw_data = pd.read_csv(f'data/raw/{instrument}_data.csv')
                    raw_data['Date'] = pd.to_datetime(raw_data['Date'])
                    raw_data['Close'] = pd.to_numeric(raw_data['Close'], errors='coerce')
                    plt.plot(raw_data['Date'], raw_data['Close'], label='Исходные данные')
                    
                    # Настройка форматирования дат
                    ax = plt.gca()
                    ax.xaxis.set_major_formatter(mdates.DateFormatter('%d %b'))
                    plt.xticks(rotation=0)
                    
                    # Настройка частоты меток
                    if self.interval in ['1m', '2m', '5m']:
                        ax.xaxis.set_major_locator(mdates.MinuteLocator(interval=30))
                    elif self.interval in ['15m', '30m']:
                        ax.xaxis.set_major_locator(mdates.HourLocator(interval=2))
                    elif self.interval == '1h':
                        ax.xaxis.set_major_locator(mdates.HourLocator(interval=4))
                    else:
                        ax.xaxis.set_major_locator(mdates.DayLocator(interval=1))
                    ax.xaxis.set_major_locator(MaxNLocator(prune='upper'))  # Ограничение количества меток
                    plt.title('Исходные данные')
                    plt.xlabel('Дата')
                    plt.ylabel('Цена закрытия')
                    plt.legend()
                    
                    # График обработанных данных
                    plt.subplot(1, 3, 2)
                    plt.plot(data['Date'], data['Close'], label='Обработанные данные', color='orange')
                    
                    # Настройка форматирования дат
                    ax = plt.gca()
                    ax.xaxis.set_major_formatter(mdates.DateFormatter('%d %b'))
                    plt.xticks(rotation=0)
                    # Настройка частоты меток
                    if self.interval in ['1m', '2m', '5m']:
                        ax.xaxis.set_major_locator(mdates.MinuteLocator(interval=30))
                    elif self.interval in ['15m', '30m']:
                        ax.xaxis.set_major_locator(mdates.HourLocator(interval=2))
                    elif self.interval == '1h':
                        ax.xaxis.set_major_locator(mdates.HourLocator(interval=4))
                    else:
                        ax.xaxis.set_major_locator(mdates.DayLocator(interval=1))
                    ax.xaxis.set_major_locator(MaxNLocator(prune='upper'))  # Ограничение количества меток
                    plt.title('Обработанные данные')
                    plt.xlabel('Дата')
                    plt.ylabel('Цена закрытия')
                    plt.legend()
                    
                    # График нормализованных данных
                    plt.subplot(1, 3, 3)
                    plt.plot(normalized_data['Date'], normalized_data['Close_normalized'],
                             label='Нормализованные данные', color='green')
                    # Настройка форматирования дат
                    ax = plt.gca()
                    ax.xaxis.set_major_formatter(mdates.DateFormatter("%d %b"))
                    plt.xticks(rotation=0)
                    # Настройка частоты меток
                    if self.interval in ['1m', '2m', '5m']:
                        ax.xaxis.set_major_locator(mdates.MinuteLocator(interval=30))
                    elif self.interval in ['15m', '30m']:
                        ax.xaxis.set_major_locator(mdates.HourLocator(interval=2))
                    elif self.interval == '1h':
                        ax.xaxis.set_major_locator(mdates.HourLocator(interval=4))
                    else:
                        ax.xaxis.set_major_locator(mdates.DayLocator(interval=1))
                    ax.xaxis.set_major_locator(MaxNLocator(prune='upper'))  # Ограничение количества меток
                    plt.title('Нормализованные данные')
                    plt.xlabel('Дата')
                    plt.ylabel('Нормализованная цена')
                    plt.legend()
                    plt.tight_layout(h_pad=2)
                    plt.show()

            except Exception as e:
                print(f"Ошибка при нормализации данных для {instrument}: {e}")
                self.status.loc[instrument, 'Normalized'] = f'Error: {str(e)}'
                self.update_dashboard()
        # Создание сводной таблицы результатов
        print(f"\n\033[1;34mСводная таблица результатов нормализации данных \033[0m")
        results_df = pd.DataFrame(results)
        print(results_df)
        results_df.to_csv('data/normalization_results.csv', index=False)
        print(f"Сводная таблица результатов сохранена в data/normalization_results.csv")
    
    def generate_features(self):
        """
        Генерация признаков на основе нормализованных данных.
        Метод использует технические индикаторы и статистические методы для создания новых признаков,
        которые могут быть полезны для построения моделей машинного обучения.
        """
        print(f"\n\033[1;34mГенерация признаков для каждого инструмента на основе нормализованных данных \033[0m")
        
        # Сначала загрузим все нормализованные данные для расчета корреляций
        all_normalized_data = {}
        for instrument in self.instruments:
            try:
                best_method = self.best_normalization_methods.get(instrument, 'Z-Score')
                normalized_file = f'data/processed/{instrument}_{best_method}_normalized.csv'
                if os.path.exists(normalized_file):
                    data = pd.read_csv(normalized_file)
                    data['Date'] = pd.to_datetime(data['Date'])
                    all_normalized_data[instrument] = data.set_index('Date')['Close_normalized']
            except Exception as e:
                print(f"Ошибка при загрузке нормализованных данных для {instrument}: {e}")

        # Расчет корреляций между инструментами
        if len(all_normalized_data) > 1:
            combined_data = pd.DataFrame(all_normalized_data)
            correlation_matrix = combined_data.corr()
            
            # Находим пары с корреляцией > 0.7
            high_correlations = {}
            for i in correlation_matrix.index:
                for j in correlation_matrix.columns:
                    if i < j:  # Избегаем дублирования
                        corr = correlation_matrix.loc[i, j]
                        if abs(corr) > 0.7:
                            high_correlations[(i, j)] = corr
                            
        # Генерация признаков для каждого инструмента                    
        for instrument in self.instruments:
            try:
                # Используем лучший метод нормализации
                best_method = self.best_normalization_methods.get(instrument, 'Z-Score')
                normalized_file = f'data/processed/{instrument}_{best_method}_normalized.csv'
                if not os.path.exists(normalized_file):
                    print(f"Файл {normalized_file} не найден. Пропуск генерации признаков.")
                    continue
                
                data = pd.read_csv(normalized_file)
                data['Date'] = pd.to_datetime(data['Date'])
                
                # Заполняем пропуски в Close_normalized методом forward fill, затем backward fill
                data['Close_normalized'] = data['Close_normalized'].fillna(method='ffill').fillna(method='bfill')
                
                # Технические индикаторы на основе нормализованных данных
                data['EMA_20'] = ta.trend.ema_indicator(data['Close_normalized'], window=20)
                data['RSI_14'] = ta.momentum.rsi(data['Close_normalized'], window=14)
                macd = ta.trend.MACD(data['Close_normalized'])
                data['MACD'] = macd.macd()
                data['MACD_signal'] = macd.macd_signal()
                
                # Добавление корреляционных признаков
                for (instr1, instr2), corr in high_correlations.items():
                    if instrument in [instr1, instr2]:
                        other_instr = instr2 if instrument == instr1 else instr1
                        corr_name = f'Corr_{other_instr}_{abs(corr):.2f}'
                        other_data = all_normalized_data[other_instr]
                        # Ресемплируем данные к одинаковым датам
                        aligned_data = pd.DataFrame({
                            'current': data.set_index('Date')['Close_normalized'],
                            'other': other_data
                        }).fillna(method='ffill')
                        # Добавляем корреляционный признак
                        rolling_corr = aligned_data['current'].rolling(window=20).corr(aligned_data['other'])
                        data[corr_name] = rolling_corr
                        
                # Заполняем NaN в технических индикаторах
                tech_indicators = ['EMA_20', 'RSI_14', 'MACD', 'MACD_signal']
                for col in tech_indicators:
                    data[col] = data[col].fillna(method='ffill').fillna(method='bfill')
                
                # Генерация признаков с помощью TSFresh с более простыми характеристиками
                tsfresh_features = self.generate_simplified_tsfresh_features(data)
                
                # Генерация признаков с помощью FeatureTools
                featuretools_features, _ = self.generate_simplified_featuretools_features(data)
                
                # Добавляем префиксы к именам признаков для указания библиотеки
                tsfresh_features = tsfresh_features.add_prefix('TSFresh_')
                featuretools_features = featuretools_features.add_prefix('FeatureTools_')
                
                # Объединение технических индикаторов и признаков TSFresh и FeatureTools
                if not tsfresh_features.empty:
                    data = pd.concat([data, tsfresh_features], axis=1)
                if not featuretools_features.empty:
                    data = pd.concat([data, featuretools_features], axis=1)
                
                # Заполнение оставшихся NaN значений
                data = data.fillna(method='ffill').fillna(method='bfill')
                
                # Сохранение признаков
                data.to_csv(f'data/processed/{instrument}_features.csv', index=False)
                print(f"Признаки сгенерированы для {instrument} и сохранены в файл data/processed/{instrument}_features.csv")
                print("=" * 150)
                
                # Обновление статуса
                self.status.loc[instrument, 'Features Generated'] = 'Complete'
                self.update_dashboard()
            except Exception as e:
                print(f"Ошибка при генерации признаков для {instrument}: {e}")
                self.status.loc[instrument, 'Features Generated'] = f'Error: {str(e)}'
                self.update_dashboard()
    
    def generate_simplified_tsfresh_features(self, data):
        """
        Генерация упрощенного набора признаков с использованием библиотеки TSFresh.
        Метод извлекает статистические характеристики из временных рядов.
        """
        try:
            # Подготовка данных для TSFresh
            tsfresh_data = data[['Date', 'Close_normalized']].copy()
            tsfresh_data['id'] = 1
            tsfresh_data = tsfresh_data.rename(columns={'Date': 'time', 'Close_normalized': 'value'})
            
            # Определяем корректный набор характеристик для извлечения
            fc_parameters = {
                "mean": None,
                "standard_deviation": None,  # Правильное имя вместо "std"
                "maximum": None,            # Правильное имя вместо "max"
                "minimum": None,            # Правильное имя вместо "min"
                "median": None,
                "skewness": None,
                "kurtosis": None,
                "sum_values": None,         # Правильное имя вместо "sum"
                "variance": None,
                "abs_energy": None,
                "count_above_mean": None,
                "count_below_mean": None,
                "first_location_of_maximum": None,
                "first_location_of_minimum": None,
                "last_location_of_maximum": None,
                "last_location_of_minimum": None,
                "longest_strike_above_mean": None,
                "longest_strike_below_mean": None,
                "mean_abs_change": None,
                "mean_change": None,
                "percentage_of_reoccurring_values_to_all_values": None,
                "ratio_value_number_to_time_series_length": None,
            }
            
            # Извлечение признаков с помощью TSFresh
            extracted_features = extract_features(tsfresh_data, 
                                               column_id='id', 
                                               column_sort='time',
                                               default_fc_parameters=fc_parameters,
                                               n_jobs=1)
            
            # Заполнение NaN значений
            extracted_features = extracted_features.fillna(method='ffill').fillna(method='bfill')
            
            # Удаление признаков, которые содержат только NaN или inf
            extracted_features = extracted_features.replace([np.inf, -np.inf], np.nan)
            extracted_features = extracted_features.dropna(axis=1, how='all')
            
            # Если остались какие-то inf значения, заменяем их на NaN и заполняем
            extracted_features = extracted_features.replace([np.inf, -np.inf], np.nan)
            extracted_features = extracted_features.fillna(extracted_features.mean())
            
            return extracted_features
        except Exception as e:
            print(f"Ошибка при генерации TSFresh признаков: {e}")
            return pd.DataFrame()
    
    def generate_simplified_featuretools_features(self, data):
        """
        Генерация упрощенного набора признаков с использованием библиотеки FeatureTools.
        Метод создает агрегированные признаки на основе данных.
        """
        try:
            # Создание EntitySet для FeatureTools
            es = ft.EntitySet(id='financial_data')
            
            # Добавление уникального индекса
            data = data.copy()
            data['id'] = range(1, len(data) + 1)
            
            # Добавление данных в EntitySet
            es = es.add_dataframe(
                dataframe_name='data',
                dataframe=data,
                index='id',
                time_index='Date'
            )
            
            # Определение простых агрегирующих функций
            agg_primitives = ["mean", "max", "min", "std", "skew"]
            trans_primitives = ["year", "month", "day"]
            
            # Генерация признаков
            feature_matrix, feature_defs = ft.dfs(
                entityset=es,
                target_dataframe_name='data',
                agg_primitives=agg_primitives,
                trans_primitives=trans_primitives,
                max_depth=1,
                verbose=True
            )
            
            # Заполнение NaN значений
            feature_matrix = feature_matrix.fillna(method='ffill').fillna(method='bfill')
            
            return feature_matrix, feature_defs
        except Exception as e:
            print(f"Ошибка при генерации FeatureTools признаков: {e}")
            return pd.DataFrame(), []
    
    def feature_importance(self):
        """
        Оценка важности признаков для построения моделей машинного обучения.
        Метод использует различные подходы (SHAP, Permutation Importance) для определения значимости каждого признака.
        """
        print(f"\n\033[1;34mОценка важности признаков на основе сгенерированных признаков \033[0m")
    
        metrics_summary = []  # Для хранения метрик по всем инструментам
        
        for instrument in self.instruments:
            try:
                features_file = f'data/processed/{instrument}_features.csv'
                if not os.path.exists(features_file):
                    print(f"Файл {features_file} не найден. Пропуск оценки важности признаков.")
                    continue
    
                data = pd.read_csv(features_file)
                data['Date'] = pd.to_datetime(data['Date'])
    
                # Подготовка данных
                feature_columns = [col for col in data.columns if col not in ['Date', 'Close_normalized', 'id']]
                feature_columns = list(dict.fromkeys(feature_columns))
    
                # Создаем словарь для хранения оригинальных имен признаков
                original_names = {}
                for col in feature_columns:
                    if col.startswith('TSFresh_'):
                        original_names[col] = col[8:]  # Убираем 'TSFresh_'
                    elif col.startswith('FeatureTools_'):
                        original_names[col] = col[12:]  # Убираем 'FeatureTools_'
                    else:
                        original_names[col] = col
    
                if len(feature_columns) == 0:
                    print(f"Недостаточно признаков для {instrument}")
                    continue
    
                X = data[feature_columns]
                y = data['Close_normalized']
    
                # Заполнение пропущенных значений
                X = X.fillna(X.median())
                y = y.fillna(y.median())
    
                # Разделение данных с сохранением временной последовательности
                train_size = int(len(X) * 0.8)
                X_train, X_test = X[:train_size], X[train_size:]
                y_train, y_test = y[:train_size], y[train_size:]
                
                # Создание и обучение модели
                model = RandomForestRegressor(n_estimators=100, random_state=42)
                model.fit(X_train, y_train)
    
                # Получение предсказаний
                y_pred_train = model.predict(X_train)
                y_pred_test = model.predict(X_test)
    
                # Расчет метрик качества
                metrics = {
                    'Инструмент': instrument,
                    'MAE (train)': mean_absolute_error(y_train, y_pred_train),
                    'MAE (test)': mean_absolute_error(y_test, y_pred_test),
                    'MSE (train)': mean_squared_error(y_train, y_pred_train),
                    'MSE (test)': mean_squared_error(y_test, y_pred_test),
                    'RMSE (train)': np.sqrt(mean_squared_error(y_train, y_pred_train)),
                    'RMSE (test)': np.sqrt(mean_squared_error(y_test, y_pred_test)),
                    'R2 (train)': r2_score(y_train, y_pred_train),
                    'R2 (test)': r2_score(y_test, y_pred_test)
                }
                metrics_summary.append(metrics)
    
                # Расчет важности признаков различными методами
                perm_importance = permutation_importance(model, X_test, y_test, n_repeats=10, random_state=42)
                explainer = shap.TreeExplainer(model)
                shap_values = explainer.shap_values(X_test)
                feature_importances = model.feature_importances_
    
                # Создание DataFrame с важностью признаков, используя очищенные имена
                importance_df = pd.DataFrame({
                    'Признак': [original_names[col] for col in feature_columns],
                    'Permutation Importance': perm_importance.importances_mean,
                    'SHAP Importance': np.abs(shap_values).mean(axis=0),
                    'Embedded Importance': feature_importances
                })
    
                # Определение типа признака и библиотеки
                def get_feature_type(feature_name):
                    original_feature = feature_name
                    for col in feature_columns:
                        if original_names[col] == feature_name:
                            if col.startswith('TSFresh_'):
                                return 'TSFresh'
                            elif col.startswith('FeatureTools_'):
                                return 'FeatureTools'
                            elif col in ['EMA_20', 'RSI_14', 'MACD', 'MACD_signal']:
                                return 'Индикатор'
                            elif 'corr' in col.lower():
                                return 'Корреляция'
                    return 'Другой'
    
                importance_df['Тип признака'] = importance_df['Признак'].apply(get_feature_type)
    
                # Фильтрация признаков по типу (FeatureTools, TSFresh, Индикаторы, Корреляции)
                filtered_features = importance_df[importance_df['Тип признака'].isin(['FeatureTools', 'TSFresh', 'Индикатор', 'Корреляция'])]
    
                # Выбор топ-15 признаков для каждой метрики
                top_features = set()
                for column in ['Permutation Importance', 'SHAP Importance', 'Embedded Importance']:
                    top_15 = filtered_features.nlargest(15, column)['Признак']
                    top_features.update(top_15)
    
                # Фильтрация таблицы важности
                importance_df = importance_df[importance_df['Признак'].isin(top_features)]
    
                # Сортировка по суммарной важности
                importance_df['Total Importance'] = (
                    importance_df['Permutation Importance'] + 
                    importance_df['SHAP Importance'] + 
                    importance_df['Embedded Importance']
                )
                importance_df = importance_df.sort_values('Total Importance', ascending=False)
    
                # Сохранение результатов
                importance_df.to_csv(f'data/processed/{instrument}_feature_importance.csv', index=False)
                print(f"Оценки признаков для {instrument} сохранены в файл data/processed/{instrument}_feature_importance.csv")
                print("=" * 150)
    
                # Обновление статуса
                self.status.loc[instrument, 'Feature Importance'] = 'Complete'
                self.update_dashboard()  

                # Визуализация результатов
                plt.figure(figsize=(20, 12))

                # График 1: Важность признаков
                plt.subplot(2, 2, 1)
                sns.heatmap(
                    importance_df.set_index(['Признак', 'Тип признака'])[
                        ['Permutation Importance', 'SHAP Importance', 'Embedded Importance']
                    ],
                    annot=True, 
                    cmap='coolwarm', 
                    fmt='.4f'
                )
                plt.title(f'Важность признаков для {instrument}')
    
                # График 2: Предсказанные vs Реальные значения (Training)
                plt.subplot(2, 2, 2)
                plt.scatter(y_train, y_pred_train, alpha=0.5, label='Training data')
                plt.plot([y_train.min(), y_train.max()], [y_train.min(), y_train.max()], 'r--', lw=2)
                plt.xlabel('Реальные значения')
                plt.ylabel('Предсказанные значения')
                plt.title(f'Предсказание vs Реальность (Training)\nR² = {r2_score(y_train, y_pred_train):.4f}')
                plt.legend()
    
                # График 3: Предсказанные vs Реальные значения (Test)
                plt.subplot(2, 2, 3)
                plt.scatter(y_test, y_pred_test, alpha=0.5, label='Test data')
                plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
                plt.xlabel('Реальные значения')
                plt.ylabel('Предсказанные значения')
                plt.title(f'Предсказание vs Реальность (Test)\nR² = {r2_score(y_test, y_pred_test):.4f}')
                plt.legend()
    
                # График 4: Временной ряд с предсказаниями
                plt.subplot(2, 2, 4)
                test_dates = data['Date'].iloc[train_size:]
                plt.plot(test_dates, y_test, label='Реальные значения', alpha=0.7)
                plt.plot(test_dates, y_pred_test, label='Предсказания', alpha=0.7)
                plt.xlabel('Дата')
                plt.ylabel('Нормализованная цена')
                plt.title('Сравнение предсказаний с реальными значениями')
                plt.legend()
                plt.xticks(rotation=45)
    
                plt.tight_layout()
                plt.savefig(f'data/processed/{instrument}_prediction_analysis.png')
                plt.show()
                
                # Вывод определений признаков
                print(f"\n\033[1;34mОпределения сгенерированных признаков \033[0m")
                print("=" * 150)
                for feature_type in importance_df['Тип признака'].unique():
                    features = importance_df[importance_df['Тип признака'] == feature_type]
                    print(f"\n{feature_type}:")
                    for _, row in features.iterrows():
                        feature = row['Признак']
                        if feature_type == 'TSFresh':
                            feature_name = feature.replace('TSfresh_', '')
                            print(f"  {feature}: TSFresh - {feature_name}")
                        elif feature_type == 'FeatureTools':
                            print(f"  {feature}: FeatureTools - Агрегированный признак")
                        elif feature_type == 'Индикатор':
                            descriptions = {
                                'EMA_20': 'Экспоненциальная скользящая средняя за 20 периодов',
                                'RSI_14': 'Индекс относительной силы за 14 периодов',
                                'MACD': 'Индикатор схождения/расхождения скользящих средних',
                                'MACD_signal': 'Сигнальная линия MACD'
                            }
                            print(f"  {feature}: {descriptions.get(feature, 'Технический индикатор')}")
                        elif feature_type == 'Корреляция':
                            print(f"  {feature}: Корреляция с другими инструментами (>50%)")
                print("=" * 150)             

            except Exception as e:
                print(f"Ошибка при оценке важности признаков для {instrument}: {e}")
                self.status.loc[instrument, 'Feature Importance'] = f'Error: {str(e)}'
                self.update_dashboard()
                
        # Вывод сводной таблицы метрик по всем инструментам
        if metrics_summary:
                metrics_df = pd.DataFrame(metrics_summary)
                print(f"\n\033[1;34mСводная таблица метрик качества предсказания \033[0m")                
                print("=" * 150)
                print(metrics_df.to_string(index=False))
                print("=" * 150)
                    
                # Сохранение метрик в файл
                metrics_df.to_csv('data/processed/prediction_metrics_summary.csv', index=False)
                print("\nМетрики качества предсказания сохранены в файл data/processed/prediction_metrics_summary.csv")
    
    def run_pipeline(self):
        """
        Основной метод запуска конвейера, который последовательно выполняет все этапы обработки данных: загрузку, обработку, нормализацию, генерацию признаков и оценку их важности.
        Метод поддерживает автоматический режим обновления данных.
        """
        try:
            # Сначала запускаем конвейер обработки данных
            print(f"\n\033[1;34mЗапуск конвейера обработки и нормализации данных, генерации признаков и оценок их важности для модели машинного обучения...\033[0m")
            print("=" * 200)
            self.download_data(mode='manual')
            print("\nПостроение матрицы корреляции данных...")
            self.show_correlation_matrix()
            print("\nЗапуск обработки данных...")
            self.process_data()
            print("\nЗапуск нормализации данных...")
            self.normalize_data()
            print("\nГенерация признаков...")
            self.generate_features()
            print("\nОценка важности признаков...")
            self.feature_importance()
    
            print("\nПредварительная обработка данных завершена")
            print("=" * 200)
    
            # После завершения обработки данных предлагаем пользователю перейти в автоматический режим
            auto_mode = input("\nХотите перейти в автоматический режим обновления данных? (y/n): ").strip().lower()
            if auto_mode == 'y':
                self.check_interval = int(input("\nВведите интервал проверки наличия новых данных в минутах: "))
                self.max_checks = int(input("Введите максимальное количество проверок наличия новых данных: "))
    
                print(f"\n\033[1;34mЗапуск автоматического режима\033[0m")
                print("=" * 200)
                print(f"Интервал проверки: {self.check_interval} минут")
                print(f"Максимальное количество проверок наличия новых данных: {self.max_checks}")
    
                while self.is_running and self.current_check < self.max_checks:
                    if not self.root.winfo_exists():  # Проверка существования окна
                        break
                        
                    print(f"\nПроверка {self.current_check + 1} из {self.max_checks}")
                    has_new_data = self.download_data(mode='auto')
    
                    if has_new_data:
                        print(f"\n\033[1;34mОбнаружены новые порции данные. Запуск конвейера обработки для обновленных данных...\033[0m")
                        print("=" * 200)
                        print("\nПостроение обновленной матрицы корреляции...")
                        self.show_correlation_matrix()
                        print("\nЗапуск обработки обновленных данных...")
                        self.process_data()
                        print("\nЗапуск нормализации обновленных данных...")
                        self.normalize_data()
                        print("\nГенерация признаков для обновленных данных...")
                        self.generate_features()
                        print("\nОбновление оценки важности признаков...")
                        self.feature_importance()
                        self.current_check = 0  # Сброс счетчика при наличии новых данных
                    else:
                        self.current_check += 1
                        if self.current_check < self.max_checks:
                            self.waiting_for_data = True
                            self.ok_button.pack(side='right', padx=5)
                            try:
                                messagebox.showinfo(
                                    "Нет новых порций данных", 
                                    f"Нет новых порций данных. Ожидаю {self.check_interval} мин...."
                                )
                                if self.root.winfo_exists():  # Проверка существования окна
                                    self.root.update()
                                    time.sleep(self.check_interval * 60)
                            except tk.TclError:
                                break
                        else:
                            try:
                                messagebox.showinfo(
                                    "Автоматический режим", 
                                    "Достигнуто максимальное количество проверок наличия новых данных. Работа конвейера завершена."
                                )
                            except tk.TclError:
                                pass
                            break
    
                    if not self.is_running:
                        break
            else:
                print("\nАвтоматический режим отменен. Конвейер обработки данных остановлен.")
    
        except Exception as e:
            print(f"Ошибка при выполнении работы конвейера: {e}")
        finally:
            self.is_running = False
            self.close_dashboard()

    def on_close(self):
        """
        Обработчик закрытия окна.
        Метод корректно завершает работу конвейера и закрывает графический интерфейс.
        """
        self.is_running = False
        self.close_dashboard()
    
    def close_dashboard(self):
        """
        Метод для корректного закрытия дашборда.
        Обеспечивает завершение работы графического интерфейса без ошибок.
        """
        if self.root.winfo_exists():
            try:
                self.root.quit()
                self.root.destroy()
            except tk.TclError:
                pass
    
    def schedule_pipeline(self):
        """
        Планировщик для автоматического запуска конвейера.
        Метод управляет периодическим выполнением задач и обработкой обновлений данных.
        """
        try:
            self.run_pipeline()
        except KeyboardInterrupt:
            print("\nПринудительное завершение работы")
            self.is_running = False
            self.close_dashboard()
        except Exception as e:
            print(f"Ошибка при выполнении планировщика: {e}")
            self.is_running = False
            self.close_dashboard()
        
if __name__ == "__main__":
    # Создаем экземпляр pipeline 
    pipeline = DataPipeline()
    # Запускаем планировщик
    pipeline.schedule_pipeline()
```
