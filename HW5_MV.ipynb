```python
"""
Copyright (c) 2025 Matvei Vasetsov (Матвей Васецов). All rights reserved.
Этот код является интеллектуальной собственностью и защищен авторским правом. Любое использование, копирование или распространение без разрешения автора запрещено.
"""
import os
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import pytorch_lightning as pl
from pytorch_forecasting import TemporalFusionTransformer, TimeSeriesDataSet
from pytorch_lightning.callbacks import EarlyStopping
from pytorch_forecasting.data import NaNLabelEncoder
import optuna
import logging
import warnings
from typing import Dict, List, Tuple
import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix
from sklearn.cluster import KMeans
import shap
import yfinance as yf
import pandas_ta as ta
from statsmodels.tsa.stattools import grangercausalitytests
from collections import Counter
from sklearn.utils.class_weight import compute_class_weight
import networkx as nx
from torch.cuda.amp import autocast, GradScaler
from IPython.display import display, HTML  

"""
### Описание алгоритма торговой стратегии

#### Общая цель
Торговая стратегия реализована в Python с использованием двух моделей машинного обучения — LSTM (Long Short-Term Memory) и TFT (Temporal Fusion Transformer) — для прогнозирования движений цен активов (акции и криптовалюты). Основная цель — максимизация доходности портфеля при управлении рисками через генерацию торговых сигналов, хеджирование и оптимизацию параметров стратегии.

#### Основные компоненты
1. **Загрузка данных (`EnhancedDataLoader`)**:
   - Источник: Yahoo Finance.
   - Активы: акции (например, MSTR, MARA) и криптовалюты (например, BTC-USD, ETH-USD).
   - Период: с 5 сентября 2019 года по 25 февраля 2025 года.
   - Данные: цены открытия, максимума, минимума, закрытия и объем торгов.
   - Логика: данные кэшируются локально в CSV-файлах, при отсутствии загружаются заново.

2. **Предобработка данных (`AdvancedPreprocessor`)**:
   - Удаление выбросов: ограничение значений по межквартильному размаху (IQR).
   - Технические индикаторы: EMA (5 и 10 дней), MACD (5, 10, 9), RSI (7 дней), ATR (7 дней), лаги цен закрытия (1-5 дней).
   - Нормализация: стандартизация числовых признаков с сохранением исходных цен закрытия.
   - Дополнительно: добавление категориальных признаков (день недели, квартал).

3. **Анализ рынка (`EnhancedMarketAnalyzer`)**:
   - **Тест Грейнджера**: выявление причинно-следственных связей между активами (p-value < 0.05).
   - **Корреляция**: вычисление матриц корреляции Пирсона и Спирмена для последних 60 дней.
   - **Кластеризация**: группировка активов по корреляции методом KMeans (5 кластеров).

4. **Модели прогнозирования**:
   - **LSTM**: двунаправленная сеть с вниманием (Multihead Attention), входной размер 14 признаков, выход — 3 класса (sell, buy, hold).
   - **TFT**: модель Temporal Fusion Transformer с учетом временных рядов и категориальных данных, выход — 3 класса.
   - Целевая переменная: процентное изменение цены закрытия (>1% — buy, <-1% — sell, иначе — hold).
   - Обучение: 5 эпох с ранней остановкой, оптимизация AdamW, веса классов для балансировки.

5. **Торговая стратегия (`EnhancedTradingStrategy`)**:
   - **Инициализация**:
     - Начальный капитал: 100,000 единиц.
     - Параметры: порог уверенности, множители стоп-лосса и тейк-профита, процент риска на сделку.
     - Использует результаты анализа рынка для хеджирования.

   - **Генерация сигналов (`generate_signals`)**:
     - **Бизнес-логика сигналов**:
       1. Для каждого актива извлекаются последние данные (минимум 1 день).
       2. Модели (LSTM и TFT) прогнозируют вероятности для классов sell, buy, hold.
       3. Выбирается действие с максимальной уверенностью (softmax).
       4. Если уверенность ≥ порога (например, 0.3):
          - Buy: покупка с расчетом размера позиции.
          - Sell: продажа с расчетом размера позиции.
          - Hold: удержание без действий.
       5. **Размер позиции**:
          - Рассчитывается как `risk_amount / (ATR * stop_loss_mult)`, где `risk_amount = capital * size_percent`.
          - Ограничение: максимум 10% капитала на сделку, корректировка по риску портфеля.
       6. **Стоп-лосс и тейк-профит**:
          - Стоп-лосс: ± `stop_loss_mult * ATR`.
          - Тейк-профит: ± `take_profit_mult * ATR`.
       7. **Хеджирование**:
          - Если корреляция Спирмена < -0.6 с другим активом:
            - Buy на основном активе → Sell на хеджирующем (размер 50% от основного).
            - Sell на основном → Buy на хеджирующем.
          - Используется для снижения риска.

   - **Исполнение сделок (`execute_trades`)**:
     - **Бизнес-логика операций**:
       1. **Открытие позиции**:
          - Buy: вычитается стоимость (цена * размер + комиссия 0.1% + проскальзывание 0.05%).
          - Sell: добавляется стоимость (цена * размер - комиссия - проскальзывание).
          - Проверка: капитал должен покрывать затраты, иначе размер корректируется.
       2. **Закрытие позиции**:
          - Противоположный сигнал (buy → sell или sell → buy): закрытие текущей позиции с расчетом прибыли/убытка.
          - Стоп-лосс: цена достигает уровня стоп-лосса → закрытие с убытком.
          - Тейк-профит: цена достигает уровня тейк-профита → закрытие с прибылью.
          - Расчет: прибыль = размер * (цена выхода - цена входа) для Buy, обратное для Sell, минус комиссия.
       3. **Управление капиталом**:
          - Капитал обновляется после каждой операции.
          - Если капитал ≤ 0, новые позиции не открываются.

6. **Оценка стратегии (`EnhancedStrategyEvaluator`)**:
   - **Бэктест**:
     - Период: тренировочный (до 15 октября 2023) и валидационный (после).
     - Шаг: ежедневно генерируются сигналы и исполняются сделки.
     - Учет открытых позиций: текущая стоимость добавляется к капиталу.
   - **Метрики**:
     - Sharpe Ratio, Max Drawdown, Total Return, Profit/Loss Ratio, Win Rate, Calmar Ratio, Sortino Ratio.
     - Сравнение с Buy-and-Hold: средняя доходность активов за период.

7. **Оптимизация**:
   - Используется Optuna для максимизации Sharpe Ratio.
   - Параметры: порог уверенности (0.3-0.9), множители стоп-лосса (2-4) и тейк-профита (4-9), процент риска (0.05-0.5).
   - Итог: лучшие параметры применяются для финального бэктеста.

8. **Визуализация (`EnhancedVisualizer`)**:
   - Дашборд: кривые капитала, таблицы сделок, оптимальные параметры для LSTM и TFT.

#### Пример работы
- **День 1**:
  - Актив: BTC-USD, цена = 60,000, ATR = 1,000.
  - LSTM: Buy (уверенность 0.7), TFT: Hold.
  - Размер = (100,000 * 0.05) / (1,000 * 2) = 2.5 единиц.
  - Стоп-лосс = 60,000 - 2,000 = 58,000, Тейк-профит = 60,000 + 4,000 = 64,000.
  - Исполнение: покупка 2.5 BTC за 150,000 + 150 (комиссия) + 75 (проскальзывание), капитал = 100,000 - 150,225 = -50,225 (корректировка размера до 0.665 BTC, капитал = 60,000).
- **День 2**:
  - Цена = 65,000, достигается тейк-профит.
  - Закрытие: продажа 0.665 BTC за 43,225 - 43.23 (комиссия), прибыль = 0.665 * (65,000 - 60,000) = 3,325, итог = 60,000 + 3,281.77 = 63,281.77.

#### Вывод
Стратегия сочетает прогнозирование с помощью LSTM и TFT, управление рисками через ATR и хеджирование по корреляции, обеспечивая адаптивность к рыночным условиям. Оптимизация параметров повышает эффективность на исторических данных.
"""

# Проверка, запущен ли код в Jupyter
def is_jupyter():
    try:
        from IPython import get_ipython
        return get_ipython() is not None
    except:
        return False

# Configuration
INITIAL_CAPITAL = 100000  # Единый начальный капитал для всех методов и моделей
warnings.filterwarnings("ignore")
# Очистка файла training.log перед началом
with open("training.log", "w") as f:
    f.truncate(0)  # Обнуляем файл

logging.basicConfig(
    level=logging.INFO,
    format='%(asctime)s - %(levelname)s - %(message)s',
    handlers=[logging.FileHandler("training.log")]
)
pl.seed_everything(42)
torch.set_default_dtype(torch.float32)

class EnhancedDataLoader:
    def __init__(self, stocks: List[str], cryptos: List[str], start_date: str = "2019-09-05", end_date: str = "2025-02-25"):
        self.stocks = stocks
        self.cryptos = cryptos
        self.start_date = start_date
        self.end_date = end_date
        self.data_dir = r"C:\Users\MV\Data_raw"
        os.makedirs(self.data_dir, exist_ok=True)
        self.all_assets = stocks + cryptos

    def fetch_yahoo_data(self) -> Dict[str, pd.DataFrame]:
        data = {}
        for asset in self.all_assets:
            try:
                df = yf.download(asset, start=self.start_date, end=self.end_date)
                if not df.empty:
                    data[asset] = df[['Open', 'High', 'Low', 'Close', 'Volume']]
                    file_path = os.path.join(self.data_dir, f"{asset}.csv")
                    df.to_csv(file_path)
                    logging.info(f"Loaded and saved {asset} to {file_path}")
                else:
                    logging.warning(f"No data for {asset}")
            except Exception as e:
                logging.error(f"Error loading {asset}: {e}")
        return data

    def load_from_csv(self) -> Dict[str, pd.DataFrame]:
        data = {}
        required_columns = ['Open', 'High', 'Low', 'Close', 'Volume']
        for asset in self.all_assets:
            file_path = os.path.join(self.data_dir, f"{asset}.csv")
            if os.path.exists(file_path):
                try:
                    df = pd.read_csv(file_path, index_col=0, parse_dates=True)
                    if all(col in df.columns for col in required_columns) and not df[required_columns].isnull().all().all():
                        df = df.loc[self.start_date:self.end_date]
                        if not df.empty:
                            data[asset] = df[required_columns]
                            logging.info(f"Loaded {asset} from {file_path}")
                        else:
                            logging.warning(f"Empty data for {asset} after filtering")
                    else:
                        logging.warning(f"Invalid CSV for {asset}")
                except Exception as e:
                    logging.error(f"Error loading {asset} from {file_path}: {e}")
            else:
                logging.info(f"No CSV for {asset}")
        return data

    def load_data(self) -> Dict[str, Dict[str, pd.DataFrame]]:
        raw_data = self.load_from_csv()
        missing_assets = [asset for asset in self.all_assets if asset not in raw_data]
        if missing_assets:
            logging.info(f"Missing assets: {missing_assets}. Fetching from Yahoo Finance.")
            yahoo_data = self.fetch_yahoo_data()
            raw_data.update({asset: yahoo_data[asset] for asset in missing_assets if asset in yahoo_data})
        
        if not raw_data:
            raise ValueError("No data loaded")

        full_index = pd.date_range(start=self.start_date, end=self.end_date, freq='D')
        for asset in raw_data:
            raw_data[asset].index = pd.to_datetime(raw_data[asset].index).normalize()
            raw_data[asset] = raw_data[asset].reindex(full_index).fillna(method='ffill').fillna(0)
                
        # Преобразование дат в Timestamp для корректного сравнения
        start_date = pd.to_datetime(self.start_date)
        end_date = pd.to_datetime(self.end_date)
        
        for asset in raw_data:
            # Удаление дубликатов индексов и временной части
            raw_data[asset].index = pd.to_datetime(raw_data[asset].index).normalize()
            if not raw_data[asset].index.is_unique:
                logging.warning(f"Duplicate indices in {asset}. Keeping first.")
                raw_data[asset] = raw_data[asset][~raw_data[asset].index.duplicated(keep='first')]
            
            # Проверка охвата диапазона дат
            if raw_data[asset].index.min() > end_date or raw_data[asset].index.max() < start_date:
                logging.warning(f"Data for {asset} outside range {start_date} to {end_date}: {raw_data[asset].index.min()} to {raw_data[asset].index.max()}")
                raw_data[asset] = pd.DataFrame(
                    index=pd.date_range(start=start_date, end=end_date, freq='D'),
                    columns=['Open', 'High', 'Low', 'Close', 'Volume']
                ).fillna(0)
            else:
                # Расширение индекса до полного диапазона и заполнение пропусков
                full_index = pd.date_range(start=start_date, end=end_date, freq='D')
                raw_data[asset] = raw_data[asset].reindex(full_index).fillna(method='ffill').fillna(0)
                # Фильтрация по диапазону дат
                raw_data[asset] = raw_data[asset].loc[start_date:end_date]
                if raw_data[asset].empty:
                    logging.warning(f"No data for {asset} in range {start_date} to {end_date}")
                    raw_data[asset] = pd.DataFrame(
                        index=pd.date_range(start=start_date, end=end_date, freq='D'),
                        columns=['Open', 'High', 'Low', 'Close', 'Volume']
                    ).fillna(0)
        
        self._visualize_data_summary(raw_data)
        return {"raw": raw_data}

    def _visualize_data_summary(self, data: Dict[str, pd.DataFrame]):
        # Определяем дату разделения на тренировочный и валидационный наборы
        split_date = pd.Timestamp('2023-10-15')
        
        summary = pd.DataFrame({
            'Asset': list(data.keys()),
            'Start Date': [df.index[0].strftime('%Y-%m-%d') if not df.empty else 'N/A' for df in data.values()],
            'End Date': [df.index[-1].strftime('%Y-%m-%d') if not df.empty else 'N/A' for df in data.values()],
            'Train End Date': [df[df.index < split_date].index[-1].strftime('%Y-%m-%d') 
                              if not df[df.index < split_date].empty else 'N/A' for df in data.values()],
            'Valid Start Date': [df[df.index >= split_date].index[0].strftime('%Y-%m-%d') 
                                if not df[df.index >= split_date].empty else 'N/A' for df in data.values()],
            'First Close': [df['Close'].iloc[0] if not df.empty else np.nan for df in data.values()],
            'Last Close': [df['Close'].iloc[-1] if not df.empty else np.nan for df in data.values()],
            'Train End Price': [df[df.index < split_date]['Close'].iloc[-1] 
                               if not df[df.index < split_date].empty else np.nan for df in data.values()],
            'Valid Start Price': [df[df.index >= split_date]['Close'].iloc[0] 
                                 if not df[df.index >= split_date].empty else np.nan for df in data.values()]
        })
        
        if is_jupyter():
            display(summary)
        else:
            print(summary.to_string())
        logging.info("Data summary displayed")

class AdvancedPreprocessor:
    def __init__(self):
        self.scalers = {}

    def process(self, data: Dict[str, pd.DataFrame], all_assets: List[str] = None) -> Dict[str, pd.DataFrame]:
        processed = {}
        all_assets = all_assets or list(data.keys())
        for asset in all_assets:
            if asset not in data or data[asset].empty or 'Close' not in data[asset].columns:
                logging.warning(f"Creating placeholder for {asset}")
                df = pd.DataFrame(index=pd.date_range(start="2019-09-05", end="2025-02-25", freq='D'),
                                columns=['Open', 'High', 'Low', 'Close', 'Volume']).fillna(0)
            else:
                df = data[asset].sort_index().reindex(columns=['Open', 'High', 'Low', 'Close', 'Volume'])
            
            df = self._handle_outliers(df)
            df = self._add_technical_indicators(df)
            df = self._normalize_features(df, asset)
            df['DayOfWeek'] = df.index.dayofweek
            df['Quarter'] = df.index.quarter
            processed[asset] = df
            self._visualize_preprocessing(asset, data.get(asset, df), df)
        return processed

    def _handle_outliers(self, df: pd.DataFrame) -> pd.DataFrame:
        for col in ['Open', 'High', 'Low', 'Close', 'Volume']:
            q1, q3 = df[col].quantile([0.25, 0.75])
            iqr = q3 - q1
            df[col] = df[col].clip(q1 - 3 * iqr, q3 + 3 * iqr)
        return df.fillna(method='ffill').fillna(method='bfill')

    def _add_technical_indicators(self, df: pd.DataFrame) -> pd.DataFrame:
        df['EMA12'] = ta.ema(df['Close'], length=5)  # Reduced from 12
        df['EMA26'] = ta.ema(df['Close'], length=10)  # Reduced from 26
        df['MACD'] = ta.macd(df['Close'], fast=5, slow=10)['MACD_5_10_9']  # Adjusted periods
        df['RSI'] = ta.rsi(df['Close'], length=7)  # Reduced from 14
        df['ATR'] = ta.atr(high=df['High'], low=df['Low'], close=df['Close'], length=7)  # Reduced from 14
        for lag in range(1, 6):  # Changed to range(1, 6) to include Close_Lag_1 to Close_Lag_5
            df[f'Close_Lag_{lag}'] = df['Close'].shift(lag)
        return df.fillna(method='ffill').fillna(method='bfill')

    def _normalize_features(self, df: pd.DataFrame, asset: str) -> pd.DataFrame:
        scaler = StandardScaler()
        numeric_cols = ['Open', 'High', 'Low', 'Close', 'Volume', 'EMA12', 'EMA26', 'MACD', 'RSI', 'ATR'] + [f'Close_Lag_{lag}' for lag in range(1, 6)]
        try:
            close_original = df['Close'].copy()
            # Only normalize columns that exist in the DataFrame
            existing_cols = [col for col in numeric_cols if col in df.columns]
            if not existing_cols:
                logging.warning(f"No columns to normalize for {asset}")
                return df
            df[existing_cols] = scaler.fit_transform(df[existing_cols])
            df['Close'] = close_original  # Restore original Close prices
            self.scalers[asset] = scaler
        except Exception as e:
            logging.error(f"Normalization error for {asset}: {e}")
            raise
        return df

    def _visualize_preprocessing(self, asset: str, raw_df: pd.DataFrame, proc_df: pd.DataFrame):
        fig = make_subplots(rows=2, cols=2, subplot_titles=[
            f"{asset}: Raw Close", f"{asset}: Processed Close",
            f"{asset}: Raw Volume", f"{asset}: Processed Volume"
        ])
        fig.add_trace(go.Histogram(x=raw_df['Close'], name='Raw Close'), row=1, col=1)
        fig.add_trace(go.Histogram(x=proc_df['Close'], name='Proc Close'), row=1, col=2)
        fig.add_trace(go.Histogram(x=raw_df['Volume'], name='Raw Volume'), row=2, col=1)
        fig.add_trace(go.Histogram(x=proc_df['Volume'], name='Proc Volume'), row=2, col=2)
        fig.update_layout(title=f"Preprocessing {asset}", height=600)
        if is_jupyter():
            display(fig)
        else:
            fig.write_image(f"preprocessing_{asset}.png")
        logging.info(f"Preprocessing visualization for {asset} displayed")
        
class EnhancedMarketAnalyzer:
    def __init__(self):
        self.granger_results = {}
        self.corr_matrix_pearson = None
        self.corr_matrix_spearman = None
        self.clusters = None

    def granger_analysis(self, data: pd.DataFrame, max_lag: int = 5) -> Dict:
        results = {}
        for col1 in data.columns:
            for col2 in data.columns:
                if col1 != col2:
                    pair_data = data[[col1, col2]].dropna()
                    try:
                        test_result = grangercausalitytests(pair_data, maxlag=max_lag, verbose=False)
                        p_values = {lag: test_result[lag][0]['ssr_chi2test'][1] for lag in range(1, max_lag + 1)}
                        min_p = min(p_values.values())
                        optimal_lag = min(p_values, key=p_values.get)
                        results[f"{col1}_to_{col2}"] = {'p_value': min_p, 'lag': optimal_lag}
                    except Exception as e:
                        logging.error(f"Granger error {col1}->{col2}: {e}")
                        results[f"{col1}_to_{col2}"] = {'p_value': 1.0, 'lag': 1}
        self.granger_results = results
        self._visualize_granger()
        return results

    def correlation_analysis(self, data: pd.DataFrame, window: int = 60) -> Tuple[pd.DataFrame, pd.DataFrame]:
        if len(data) < window:
            logging.warning(f"Data length {len(data)} is less than window {window}. Using available data.")
            window = len(data)
        analysis_data = data.tail(window).dropna()
        if analysis_data.empty:
            logging.error("No valid data after dropping NaN for correlation analysis.")
            raise ValueError("Correlation analysis failed due to insufficient valid data.")
        
        analysis_data = analysis_data.fillna(0)
        
        self.corr_matrix_pearson = analysis_data.corr(method='pearson')
        self.corr_matrix_spearman = analysis_data.corr(method='spearman')
        
        return self.corr_matrix_pearson, self.corr_matrix_spearman

    def cluster_analysis(self, data: pd.DataFrame, n_clusters: int = 5) -> np.ndarray:
        corr_matrix = self.corr_matrix_spearman if self.corr_matrix_spearman is not None else data.corr(method='spearman')
        
        if corr_matrix.isnull().any().any():
            logging.info("Filling NaN in correlation matrix with 0 for clustering.")
            corr_matrix = corr_matrix.fillna(0)
        
        kmeans = KMeans(n_clusters=n_clusters, random_state=42)
        self.clusters = kmeans.fit_predict(corr_matrix.values)
        self._visualize_clusters(data)
        return self.clusters

    def _visualize_granger(self):
        G = nx.DiGraph()
        for key, value in self.granger_results.items():
            if value['p_value'] < 0.05:
                src, tgt = key.split('_to_')
                G.add_edge(src, tgt, weight=1/value['p_value'], lag=value['lag'])
        pos = nx.spring_layout(G)
        edge_trace = [go.Scatter(x=[pos[e[0]][0], pos[e[1]][0], None], y=[pos[e[0]][1], pos[e[1]][1], None], line=dict(width=1)) for e in G.edges()]
        node_trace = go.Scatter(
            x=[pos[n][0] for n in G.nodes()],
            y=[pos[n][1] for n in G.nodes()],
            mode='markers+text',
            text=[f"{n} (Lags: {', '.join([str(G.edges[(n, tgt)]['lag']) for tgt in G.successors(n)])})" 
                  if list(G.successors(n)) 
                  else f"{n} (No outgoing edges)" 
                  for n in G.nodes()]
        )
        fig = go.Figure(data=edge_trace + [node_trace], layout=go.Layout(title="Granger Causality Network (p<0.05)"))
        if is_jupyter():
            display(fig)
        else:
            fig.write_image("granger_causality.png")
        logging.info("Granger causality graph displayed")

    def _visualize_clusters(self, data: pd.DataFrame):
        sorted_assets = [self.corr_matrix_spearman.columns[i] for i in np.argsort(self.clusters)]
        sorted_corr = self.corr_matrix_spearman.loc[sorted_assets, sorted_assets]
        fig = px.imshow(sorted_corr, color_continuous_scale='RdBu_r', zmin=-1, zmax=1, text_auto=".2f")
        fig.update_layout(title="Clustered Spearman Correlation Matrix")
        if is_jupyter():
            display(fig)
        else:
            fig.write_image("correlation_clusters.png")
        logging.info("Correlation clusters displayed")

class EnhancedLSTMModel(pl.LightningModule):
    def __init__(self, input_size: int = 14, hidden_size: int = 128, num_layers: int = 2, dropout: float = 0.2, attention_heads: int = 4, class_weights=None):
        super().__init__()
        self.save_hyperparameters()
        self.input_size = input_size
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers, batch_first=True, dropout=dropout if num_layers > 1 else 0, bidirectional=True)
        self.attention = nn.MultiheadAttention(embed_dim=hidden_size * 2, num_heads=attention_heads)
        self.residual_conn = nn.Linear(input_size, hidden_size * 2)
        self.dropout = nn.Dropout(dropout)
        self.fc1 = nn.Linear(hidden_size * 2, hidden_size)
        self.fc2 = nn.Linear(hidden_size, 3)
        self.norm1 = nn.LayerNorm(hidden_size * 2)
        self.norm2 = nn.LayerNorm(hidden_size)
        self.criterion = nn.CrossEntropyLoss(weight=class_weights)
        self.scaler = GradScaler() if torch.cuda.is_available() else None

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        if x.size(2) != self.input_size:
            logging.warning(f"Input size mismatch: expected {self.input_size}, got {x.size(2)}. Adjusting input.")
            # Pad or truncate the input to match expected size
            if x.size(2) < self.input_size:
                padding = torch.zeros(x.size(0), x.size(1), self.input_size - x.size(2), device=x.device)
                x = torch.cat([x, padding], dim=2)
            elif x.size(2) > self.input_size:
                x = x[:, :, :self.input_size]
        if torch.cuda.is_available() and self.scaler:
            with autocast():
                residual = self.residual_conn(x)
                lstm_out, _ = self.lstm(x)
                lstm_out = lstm_out.transpose(0, 1)
                attn_out, _ = self.attention(lstm_out, lstm_out, lstm_out)
                attn_out = attn_out.transpose(0, 1)
                out = self.norm1(attn_out + residual)
                out = torch.relu(self.fc1(out[:, -1, :]))
                out = self.dropout(out)
                out = self.norm2(out)
                return self.fc2(out)
        else:
            residual = self.residual_conn(x)
            lstm_out, _ = self.lstm(x)
            lstm_out = lstm_out.transpose(0, 1)
            attn_out, _ = self.attention(lstm_out, lstm_out, lstm_out)
            attn_out = attn_out.transpose(0, 1)
            out = self.norm1(attn_out + residual)
            out = torch.relu(self.fc1(out[:, -1, :]))
            out = self.dropout(out)
            out = self.norm2(out)
            return self.fc2(out)

    def training_step(self, batch, batch_idx):
        x, y_tuple = batch
        y = y_tuple[0].squeeze()
        y_hat = self(x['encoder_cont'])
        loss = self.criterion(y_hat, y.long())
        self.log('train_loss', loss, on_epoch=True, on_step=False)
        return loss

    def validation_step(self, batch, batch_idx):
        x, y_tuple = batch
        y = y_tuple[0].flatten()
        y_hat = self(x['encoder_cont'])
        loss = self.criterion(y_hat, y.long())
        preds = torch.argmax(y_hat, dim=1).cpu().numpy()
        targets = y.cpu().numpy()
        probs = torch.softmax(y_hat, dim=1).cpu().numpy()
        acc = accuracy_score(targets, preds)
        precision = precision_score(targets, preds, average='weighted', zero_division=0)
        recall = recall_score(targets, preds, average='weighted', zero_division=0)
        f1 = f1_score(targets, preds, average='weighted', zero_division=0)
        
        unique_classes = np.unique(targets)
        logging.info(f"Unique classes in targets (LSTM): {unique_classes}")
        if len(unique_classes) > 1:
            roc_auc = roc_auc_score(targets, probs, multi_class='ovr', labels=[0, 1, 2])
        else:
            roc_auc = 0.0
        
        self.log('val_loss', loss, on_epoch=True, on_step=False)
        self.log('val_accuracy', acc, on_epoch=True, on_step=False)
        self.log('val_precision', precision, on_epoch=True, on_step=False)
        self.log('val_recall', recall, on_epoch=True, on_step=False)
        self.log('val_f1', f1, on_epoch=True, on_step=False)
        self.log('val_roc_auc', roc_auc, on_epoch=True, on_step=False)
        return {'val_loss': loss, 'preds': y_hat, 'targets': y}

    def configure_optimizers(self):
        optimizer = torch.optim.AdamW(self.parameters(), lr=0.0001, weight_decay=0.01)
        scheduler = torch.optim.lr_scheduler.OneCycleLR(optimizer, max_lr=0.001, epochs=5, steps_per_epoch=2)
        return {"optimizer": optimizer, "lr_scheduler": scheduler}

class TemporalFusionTransformerModel(pl.LightningModule):
    def __init__(self, dataset: TimeSeriesDataSet, class_weights=None):
        super().__init__()
        logging.info("Initializing TFT Model")
        self.dataset = dataset
        self.tft = TemporalFusionTransformer.from_dataset(
            dataset,
            hidden_size=256,
            attention_head_size=8,
            dropout=0.3,
            lstm_layers=3,
            hidden_continuous_size=256,
            output_size=3
        )
        self.criterion = nn.CrossEntropyLoss(weight=class_weights)
        self.scaler = GradScaler() if torch.cuda.is_available() else None

    def forward(self, x):
        if torch.cuda.is_available() and self.scaler:
            with autocast():
                output = self.tft(x)
                predictions = output.prediction.squeeze(1) if output.prediction.dim() == 3 else output.prediction
                return predictions
        else:
            output = self.tft(x)
            predictions = output.prediction.squeeze(1) if output.prediction.dim() == 3 else output.prediction
            return predictions

    def training_step(self, batch, batch_idx):
        x, y_tuple = batch
        y = y_tuple[0].squeeze()
        y_hat = self(x)
        loss = self.criterion(y_hat, y.long())
        self.log('train_loss', loss, on_epoch=True, on_step=False)
        return loss

    def validation_step(self, batch, batch_idx):
        x, y_tuple = batch
        y = y_tuple[0].flatten()
        y_hat = self(x)
        loss = self.criterion(y_hat, y.long())
        preds = torch.argmax(y_hat, dim=1).cpu().numpy()
        targets = y.cpu().numpy()
        probs = torch.softmax(y_hat, dim=1).cpu().numpy()
        acc = accuracy_score(targets, preds)
        precision = precision_score(targets, preds, average='weighted', zero_division=0)
        recall = recall_score(targets, preds, average='weighted', zero_division=0)
        f1 = f1_score(targets, preds, average='weighted', zero_division=0)
        
        unique_classes = np.unique(targets)
        logging.info(f"Unique classes in targets (TFT): {unique_classes}")
        if len(unique_classes) > 1:
            roc_auc = roc_auc_score(targets, probs, multi_class='ovr', labels=[0, 1, 2])
        else:
            roc_auc = 0.0
        
        self.log('val_loss', loss, on_epoch=True, on_step=False)
        self.log('val_accuracy', acc, on_epoch=True, on_step=False)
        self.log('val_precision', precision, on_epoch=True, on_step=False)
        self.log('val_recall', recall, on_epoch=True, on_step=False)
        self.log('val_f1', f1, on_epoch=True, on_step=False)
        self.log('val_roc_auc', roc_auc, on_epoch=True, on_step=False)
        return {'val_loss': loss, 'preds': y_hat, 'targets': y}
        
    def configure_optimizers(self):
        optimizer = torch.optim.AdamW(self.parameters(), lr=0.0001, weight_decay=0.01)
        scheduler = torch.optim.lr_scheduler.CosineAnnealingWarmRestarts(optimizer, T_0=10, T_mult=1)
        return {"optimizer": optimizer, "lr_scheduler": scheduler}

class TimeSeriesDataModule(pl.LightningDataModule):
    def __init__(self, data: pd.DataFrame, batch_size: int = 64, all_assets: List[str] = None):
        super().__init__()
        self.data = data
        self.batch_size = batch_size
        self.all_assets = all_assets
        self.asset_encoder = None

    def setup(self, stage=None):
        logging.info(f"Setting up TimeSeriesDataModule with data size: {len(self.data)}")
        if not self.data.index.is_unique:
            logging.info("Removing duplicate indices.")
            self.data = self.data[~self.data.index.duplicated(keep='first')]
        
        # Удаление временной части из индексов
        self.data.index = pd.to_datetime(self.data.index).normalize()
        if not self.data.index.is_unique:
            logging.info("Removing duplicate dates after normalization.")
            self.data = self.data[~self.data.index.duplicated(keep='first')]
        
        if self.data.empty:
            raise ValueError("Input data is empty")
        
        self.data['time_idx'] = self.data.groupby('asset').cumcount()
        self.data['target'] = self.data['Close'].pct_change().apply(lambda x: 1 if x > 0.01 else (0 if x < -0.01 else 2)).fillna(2).astype(int)
        
        # Используем pd.Timestamp для корректного сравнения
        split_date = pd.Timestamp('2023-10-15')
        train_data = self.data[self.data.index < split_date]
        valid_data = self.data[self.data.index >= split_date]
        
        if train_data.empty or valid_data.empty:
            logging.error(f"Train data size: {len(train_data)}, Valid data size: {len(valid_data)}")
            # Создаем минимальные данные, если один из наборов пуст
            full_index = pd.date_range(start=self.data.index.min(), end=self.data.index.max(), freq='D')
            self.data = self.data.reindex(full_index).fillna(method='ffill').fillna(0)
            train_data = self.data[self.data.index < split_date]
            valid_data = self.data[self.data.index >= split_date]
            logging.info(f"Recreated datasets - Train size: {len(train_data)}, Valid size: {len(valid_data)}")
        
        logging.info(f"Train data size: {len(train_data)}, Validation data size: {len(valid_data)}")
        
        # Define the features and ensure they exist in the data
        time_varying_unknown_reals = ["Close", "Volume", "EMA12", "EMA26", "MACD", "RSI", "ATR"] + [f'Close_Lag_{lag}' for lag in range(1, 6)]
        # Filter out any features that don't exist in the data
        time_varying_unknown_reals = [col for col in time_varying_unknown_reals if col in train_data.columns]
        if not time_varying_unknown_reals:
            logging.error("No time-varying unknown reals available after filtering")
            raise ValueError("No valid features for TimeSeriesDataSet")

        self.train_dataset = TimeSeriesDataSet(
            train_data,
            time_idx="time_idx",
            target="target",
            group_ids=["asset"],
            max_encoder_length=5,
            max_prediction_length=1,
            time_varying_unknown_reals=time_varying_unknown_reals,
            time_varying_known_reals=["DayOfWeek", "Quarter"],
            static_categoricals=["asset"],
            categorical_encoders={"asset": NaNLabelEncoder().fit(self.all_assets)},
            allow_missing_timesteps=True
        )
        self.val_dataset = TimeSeriesDataSet.from_dataset(
            self.train_dataset,
            valid_data,
            predict=True,
            stop_randomization=True
        )
        self.asset_encoder = {asset: i for i, asset in enumerate(self.all_assets)}

    def train_dataloader(self):
        return self.train_dataset.to_dataloader(train=True, batch_size=self.batch_size)

    def val_dataloader(self):
        return self.val_dataset.to_dataloader(train=False, batch_size=self.batch_size)

class EnhancedTradingStrategy:
    def __init__(self, lstm_model, tft_model, config: Dict, asset_encoder: Dict, granger_results: Dict, corr_data: Tuple[pd.DataFrame, pd.DataFrame], clusters: np.ndarray):
        self.lstm_model = lstm_model
        self.tft_model = tft_model
        self.config = config
        self.positions = {}
        self.capital = INITIAL_CAPITAL
        self.asset_encoder = asset_encoder or {}
        self.assets = config.get('assets', [])
        self.asset_to_idx = {asset: idx for idx, asset in enumerate(self.assets)} if not asset_encoder else asset_encoder
        if self.lstm_model is None:
            raise ValueError("LSTM model cannot be None")
        logging.info(f"LSTM model initialized with input_size={self.lstm_model.hparams.input_size}")
        if self.tft_model is None:
            raise ValueError("TFT model cannot be None")
        logging.info("TFT model initialized successfully")
        self.max_asset_idx = self.tft_model.tft.input_embeddings['asset'].weight.size(0) - 1
        self.granger_results = granger_results or {}
        self.corr_pearson, self.corr_spearman = corr_data if corr_data else (None, None)
        self.clusters = clusters if clusters is not None else np.array([])
        self.cluster_map = {asset: cluster for asset, cluster in zip(self.assets, self.clusters)} if self.clusters.size > 0 else {}
        self.commission_rate = 0.001
        self.slippage_rate = 0.0005

    def _calculate_position_size(self, atr: float, market_data: Dict[str, pd.DataFrame] = None, price: float = None) -> float:
        try:
            # Risk amount per trade as a percentage of capital
            risk_amount = self.capital * self.config['size_percent']
            # Cap the position size based on available capital (max 10% of capital per trade)
            max_size_by_capital = (self.capital * 0.1) / price if price > 0 else 0
            # Base position size based on ATR and stop-loss multiplier
            base_size = risk_amount / (atr * self.config['stop_loss_mult']) if atr != 0 else 0
    
            # Adjust for portfolio risk (limit total risk exposure)
            portfolio_risk = sum(
                pos['size'] * market_data[pos['asset']]['ATR'].iloc[-1] / self.capital 
                for pos in self.positions.values() if 'asset' in pos and pos['asset'] in market_data
            ) if market_data is not None and hasattr(self, 'positions') else 0
            risk_adjustment = max(1.0 - portfolio_risk / 0.3, 0.1)
    
            # Calculate the adjusted size, capped by max_size_by_capital
            size = min(base_size * risk_adjustment, max_size_by_capital)
            # Ensure the trade cost does not exceed available capital
            if price and size * price > self.capital:
                size = self.capital / price if price > 0 else 0
    
            # Additional safety: cap the size to a reasonable maximum (e.g., 1000 units)
            size = min(size, 1000.0)
    
            logging.debug(f"Position size for ATR {atr:.8f}: Base {base_size:.8f}, Adjusted {size:.8f}, Risk Adj {risk_adjustment:.8f}")
            return max(size, 0)
        except Exception as e:
            logging.error(f"Error in position size calculation: {e}")
            return 0

    def generate_signals(self, market_data: Dict[str, pd.DataFrame]) -> Dict[str, Dict[str, Dict]]:
        signals = {'lstm': {}, 'tft': {}}
        feature_cols = ['Close', 'Volume', 'EMA12', 'EMA26', 'MACD', 'RSI', 'ATR', 'DayOfWeek', 'Quarter'] + [f'Close_Lag_{lag}' for lag in range(1, 6)]
        batch_x = []
        batch_x_dict = []
        batch_assets = []
        min_length = 1
        
        for asset, df in market_data.items():
            if asset not in self.asset_to_idx:
                logging.warning(f"Asset {asset} not in asset_to_idx. Skipping.")
                continue
            try:
                existing_cols = [col for col in feature_cols if col in df.columns]
                if not existing_cols:
                    logging.warning(f"No valid features for {asset}")
                    continue
                features = df[existing_cols].tail(min_length).dropna()
                if features.empty or len(features) < min_length:
                    logging.warning(f"Insufficient data for {asset}: {len(features)} rows, required {min_length}")
                    continue
                x = torch.tensor(features.values, dtype=torch.float32).unsqueeze(0)
                asset_idx = min(self.asset_to_idx[asset], self.max_asset_idx)
                x_dict = {
                    "encoder_cont": x,
                    "encoder_lengths": torch.full((1,), min_length, dtype=torch.long),
                    "decoder_cont": torch.zeros((1, 1, len(existing_cols)), dtype=torch.float32),
                    "decoder_lengths": torch.ones((1,), dtype=torch.long),
                    "encoder_cat": torch.full((1, min_length, 1), asset_idx, dtype=torch.long),
                    "decoder_cat": torch.full((1, 1, 1), asset_idx, dtype=torch.long),
                    "target_scale": torch.tensor([[0.0, 1.0]], dtype=torch.float32),
                }
                batch_x.append(x)
                batch_x_dict.append(x_dict)
                batch_assets.append(asset)
            except Exception as e:
                logging.error(f"Error preparing data for {asset}: {e}")
        
        if not batch_x:
            logging.warning("No valid batches prepared. Generating minimal signals for all assets.")
            for asset in market_data.keys():
                atr = market_data[asset]['ATR'].iloc[-1] if 'ATR' in market_data[asset].columns else 0.01
                price = market_data[asset]['Close'].iloc[-1]
                size = min(0.1, self.capital / price) if price > 0 else 0.1
                for model in ['lstm', 'tft']:
                    signals[model][asset] = {
                        'action': 'hold',
                        'confidence': 0.2,
                        'size': size,
                        'stop_loss': 0,
                        'take_profit': 0
                    }
            return signals
        
        with torch.no_grad():
            batch_x_tensor = torch.cat(batch_x, dim=0)
            lstm_preds = torch.softmax(self.lstm_model(batch_x_tensor), dim=1).cpu().numpy()
            batch_x_dict_combined = {k: torch.cat([x[k] for x in batch_x_dict], dim=0) for k in batch_x_dict[0]}
            tft_preds = torch.softmax(self.tft_model(batch_x_dict_combined), dim=1).cpu().numpy()
            
            # Базовые сигналы от моделей
            base_signals = {'lstm': {}, 'tft': {}}
            for asset, lstm_pred, tft_pred in zip(batch_assets, lstm_preds, tft_preds):
                atr = market_data[asset]['ATR'].iloc[-1] if 'ATR' in market_data[asset].columns else 0.01
                price = market_data[asset]['Close'].iloc[-1]
                size = self._calculate_position_size(atr, market_data, price)
                
                # LSTM сигналы
                lstm_confidence = lstm_pred.max()
                lstm_action_idx = np.argmax(lstm_pred)
                lstm_action = ['sell', 'buy', 'hold'][lstm_action_idx]
                if size > 0 and lstm_confidence >= self.config['confidence_threshold']:
                    stop_loss = self.config['stop_loss_mult'] * atr if lstm_action == 'sell' else -self.config['stop_loss_mult'] * atr
                    take_profit = -self.config['take_profit_mult'] * atr if lstm_action == 'sell' else self.config['take_profit_mult'] * atr
                    
                    base_signals['lstm'][asset] = {
                        'action': lstm_action,
                        'confidence': lstm_confidence,
                        'size': size,
                        'stop_loss': stop_loss,
                        'take_profit': take_profit
                    }
                else:
                    base_signals['lstm'][asset] = {
                        'action': 'hold',
                        'confidence': lstm_confidence,
                        'size': size,
                        'stop_loss': 0,
                        'take_profit': 0
                    }
                
                # TFT сигналы
                tft_confidence = tft_pred.max()
                tft_action_idx = np.argmax(tft_pred)
                tft_action = ['sell', 'buy', 'hold'][tft_action_idx]
                if size > 0 and tft_confidence >= self.config['confidence_threshold']:
                    stop_loss = self.config['stop_loss_mult'] * atr if tft_action == 'sell' else -self.config['stop_loss_mult'] * atr
                    take_profit = -self.config['take_profit_mult'] * atr if tft_action == 'sell' else self.config['take_profit_mult'] * atr
                    
                    base_signals['tft'][asset] = {
                        'action': tft_action,
                        'confidence': tft_confidence,
                        'size': size,
                        'stop_loss': stop_loss,
                        'take_profit': take_profit
                    }
                else:
                    base_signals['tft'][asset] = {
                        'action': 'hold',
                        'confidence': tft_confidence,
                        'size': size,
                        'stop_loss': 0,
                        'take_profit': 0
                    }
            
            # Применение хеджирования через коррелированные активы
            if self.corr_spearman is not None:
                for model in ['lstm', 'tft']:
                    for asset in batch_assets:
                        signals[model][asset] = base_signals[model][asset].copy()
                        if asset in self.corr_spearman.index:
                            corr_row = self.corr_spearman.loc[asset]
                            hedge_assets = corr_row[corr_row < -0.6].index.tolist()
                            if hedge_assets:
                                for hedge_asset in hedge_assets:
                                    if hedge_asset in market_data and hedge_asset in self.asset_to_idx:
                                        hedge_atr = market_data[hedge_asset]['ATR'].iloc[-1] if 'ATR' in market_data[hedge_asset].columns else 0.01
                                        hedge_price = market_data[hedge_asset]['Close'].iloc[-1]
                                        hedge_size = self._calculate_position_size(hedge_atr, market_data, hedge_price) * 0.5
                                        
                                        if base_signals[model][asset]['action'] == 'buy' and base_signals[model][asset]['confidence'] >= self.config['confidence_threshold']:
                                            signals[model][hedge_asset] = {
                                                'action': 'sell',
                                                'confidence': base_signals[model][asset]['confidence'],
                                                'size': hedge_size,
                                                'stop_loss': self.config['stop_loss_mult'] * hedge_atr,
                                                'take_profit': -self.config['take_profit_mult'] * hedge_atr
                                            }
                                            logging.info(f"Hedging {asset} ({model}) buy with sell on {hedge_asset}, correlation={corr_row[hedge_asset]:.2f}")
                                        elif base_signals[model][asset]['action'] == 'sell' and base_signals[model][asset]['confidence'] >= self.config['confidence_threshold']:
                                            signals[model][hedge_asset] = {
                                                'action': 'buy',
                                                'confidence': base_signals[model][asset]['confidence'],
                                                'size': hedge_size,
                                                'stop_loss': -self.config['stop_loss_mult'] * hedge_atr,
                                                'take_profit': self.config['take_profit_mult'] * hedge_atr
                                            }
                                            logging.info(f"Hedging {asset} ({model}) sell with buy on {hedge_asset}, correlation={corr_row[hedge_asset]:.2f}")
            
            for model in ['lstm', 'tft']:
                for asset in market_data.keys():
                    if asset not in signals[model]:
                        signals[model][asset] = base_signals[model].get(asset, {
                            'action': 'hold',
                            'confidence': 0.3,
                            'size': self._calculate_position_size(
                                market_data[asset]['ATR'].iloc[-1] if 'ATR' in market_data[asset].columns else 0.01,
                                market_data,
                                market_data[asset]['Close'].iloc[-1]
                            ),
                            'stop_loss': 0,
                            'take_profit': 0
                        })
        
        return signals
        
    def execute_trades(self, signals: Dict[str, Dict], market_data: Dict[str, pd.DataFrame], dataset_type: str = 'train') -> Dict[str, List[Dict]]:
        trades = {'lstm': [], 'tft': []}
        for model in ['lstm', 'tft']:
            for asset, signal in signals[model].items():
                try:
                    price = market_data[asset]['Close'].iloc[-1]
                    trade_date = market_data[asset].index[-1].strftime('%Y-%m-%d')
                    execution_price = price * (1 + self.slippage_rate if signal['action'] == 'buy' else 1 - self.slippage_rate)
                    trade_cost = signal['size'] * execution_price
                    commission = trade_cost * self.commission_rate
                    pos_key = f"{model}_{asset}"
    
                    # Проверка отрицательного капитала перед открытием позиции
                    if self.capital <= 0 and signal['action'] in ['buy', 'sell']:
                        logging.warning(f"Cannot open {signal['action']} position for {asset} ({model}) on {trade_date}: Capital is {self.capital:.2f}")
                        continue
    
                    if signal['action'] in ['buy', 'sell'] and trade_cost + commission > self.capital:
                        signal['size'] = (self.capital - commission) / execution_price if execution_price > 0 else signal['size']
                        trade_cost = signal['size'] * execution_price
                        commission = trade_cost * self.commission_rate
    
                    trade = {
                        'asset': asset,
                        'action': signal['action'],
                        'size': signal['size'],
                        'entry_price': execution_price,
                        'stop_loss': execution_price + signal['stop_loss'],
                        'take_profit': execution_price + signal['take_profit'],
                        'entry_timestamp': market_data[asset].index[-1],
                        'confidence': signal['confidence'],
                        'model': model,
                        'commission': commission
                    }
    
                    # Обработка противоположных сделок
                    if pos_key in self.positions:
                        current_pos = self.positions[pos_key]
                        if current_pos['action'] != signal['action']:
                            if current_pos['action'] == 'buy':
                                profit = current_pos['size'] * (execution_price - current_pos['entry_price'])
                                exit_commission = (current_pos['size'] * execution_price) * self.commission_rate
                                net_result = profit - exit_commission
                                self.capital += (current_pos['size'] * execution_price) + net_result
                            else:  # sell
                                profit = current_pos['size'] * (current_pos['entry_price'] - execution_price)
                                exit_commission = (current_pos['size'] * execution_price) * self.commission_rate
                                net_result = profit - exit_commission
                                self.capital -= (current_pos['size'] * execution_price) - net_result
    
                            current_pos.update({
                                'exit_price': execution_price,
                                'exit_timestamp': market_data[asset].index[-1],
                                'close_reason': 'opposite_trade',
                                'profit_loss': net_result,
                                'exit_commission': exit_commission
                            })
                            trades[model].append(current_pos)
                            logging.info(f"Closed {current_pos['action']} position for {asset} ({model}) on {trade_date}: "
                                       f"Size={current_pos['size']:.4f}, ExitPrice={execution_price:.4f}, "
                                       f"Reason=opposite_trade, Profit/Loss={net_result:.4f}")
                            del self.positions[pos_key]
    
                            # Открытие новой позиции (если капитал положительный)
                            if signal['size'] > 0 and self.capital > 0:
                                trade_cost = signal['size'] * execution_price
                                commission = trade_cost * self.commission_rate
                                trade['commission'] = commission
                                if signal['action'] == 'buy':
                                    self.capital -= trade_cost + commission
                                else:
                                    self.capital += trade_cost - commission
                                self.positions[pos_key] = trade
                                trades[model].append(trade)
                                logging.info(f"Opened {signal['action']} position for {asset} ({model}) on {trade_date}: "
                                           f"Size={trade['size']:.4f}, Price={execution_price:.4f}")
                            continue
    
                    # Открытие новой позиции
                    if signal['action'] == 'buy' and self.capital > trade_cost + commission:
                        self.capital -= trade_cost + commission
                        self.positions[pos_key] = trade
                        trades[model].append(trade)
                        logging.info(f"Opened long position for {asset} ({model}) on {trade_date}: "
                                   f"Size={trade['size']:.4f}, Price={execution_price:.4f}")
                    elif signal['action'] == 'sell' and self.capital > 0:
                        self.capital += trade_cost - commission
                        self.positions[pos_key] = trade
                        trades[model].append(trade)
                        logging.info(f"Opened short position for {asset} ({model}) on {trade_date}: "
                                   f"Size={trade['size']:.4f}, Price={execution_price:.4f}")
    
                    # Закрытие позиций по стоп-лоссу или тейк-профиту
                    if pos_key in self.positions:
                        pos = self.positions[pos_key]
                        current_price = market_data[asset]['Close'].iloc[-1]
                        execution_price = current_price * (1 - self.slippage_rate if pos['action'] == 'buy' else 1 + self.slippage_rate)
    
                        if pos['action'] == 'buy':
                            if current_price >= pos['take_profit'] or current_price <= pos['stop_loss']:
                                profit = pos['size'] * (execution_price - pos['entry_price'])
                                exit_commission = (pos['size'] * execution_price) * self.commission_rate
                                net_result = profit - exit_commission
                                self.capital += (pos['size'] * execution_price) + net_result
                                close_reason = 'take_profit' if current_price >= pos['take_profit'] else 'stop_loss'
                                pos.update({
                                    'exit_price': execution_price,
                                    'exit_timestamp': market_data[asset].index[-1],
                                    'close_reason': close_reason,
                                    'profit_loss': net_result,
                                    'exit_commission': exit_commission
                                })
                                trades[model].append(pos)
                                logging.info(f"Closed long position for {asset} ({model}) on {trade_date}: "
                                           f"Size={pos['size']:.4f}, ExitPrice={execution_price:.4f}, "
                                           f"Reason={close_reason}, Profit/Loss={net_result:.4f}")
                                del self.positions[pos_key]
                        elif pos['action'] == 'sell':
                            if current_price <= pos['take_profit'] or current_price >= pos['stop_loss']:
                                profit = pos['size'] * (pos['entry_price'] - execution_price)
                                exit_commission = (pos['size'] * execution_price) * self.commission_rate
                                net_result = profit - exit_commission
                                self.capital -= (pos['size'] * execution_price) - net_result
                                close_reason = 'take_profit' if current_price <= pos['take_profit'] else 'stop_loss'
                                pos.update({
                                    'exit_price': execution_price,
                                    'exit_timestamp': market_data[asset].index[-1],
                                    'close_reason': close_reason,
                                    'profit_loss': net_result,
                                    'exit_commission': exit_commission
                                })
                                trades[model].append(pos)
                                logging.info(f"Closed short position for {asset} ({model}) on {trade_date}: "
                                           f"Size={pos['size']:.4f}, ExitPrice={execution_price:.4f}, "
                                           f"Reason={close_reason}, Profit/Loss={net_result:.4f}")
                                del self.positions[pos_key]
    
                except Exception as e:
                    logging.error(f"Error executing trade for {asset} ({model}): {e}")
    
        return {'trades': trades, 'capital': self.capital}

class EnhancedStrategyEvaluator:
    def __init__(self, config: Dict):
        self.config = config
        self.max_portfolio_risk = 0.5
        self.commission_rate = 0.001  # 0.1% commission
        self.slippage_rate = 0.0005  # 0.05% slippage

    def evaluate(self, strategy: EnhancedTradingStrategy, market_data: Dict[str, pd.DataFrame], dataset: TimeSeriesDataSet, 
                 is_train: bool = True, max_days: int = 1000, model_type: str = 'lstm') -> Dict:
        logging.info(f"Starting evaluation (is_train={is_train}, model_type={model_type})")
        
        self.capital = INITIAL_CAPITAL
        strategy.capital = INITIAL_CAPITAL
        strategy.positions = {}
        
        results = {'trades': {model_type: []}, 'equity_curve': [], 'timestamps': []}
        open_positions = {}
        
        trading_days = list(self._generate_trading_days(market_data))
        if len(trading_days) < 5:
            logging.warning(f"Insufficient trading days: {len(trading_days)}")
            return {'metrics': {}, 'results': results}
        
        initial_prices = {asset: df['Close'].iloc[0] for asset, df in market_data.items()}
        final_prices = {asset: df['Close'].iloc[-1] for asset, df in market_data.items()}
        total_buy_and_hold_return = sum((final_prices[asset] - initial_prices[asset]) / initial_prices[asset] 
                                        for asset in initial_prices if initial_prices[asset] != 0) * 100 / len(initial_prices)
        
        split_date = pd.Timestamp('2023-10-15')
        first_trading_day = trading_days[0][0]
        
        results['equity_curve'].append(INITIAL_CAPITAL)
        results['timestamps'].append(first_trading_day if is_train else split_date)
        
        valid_days_processed = 0
        for i, (date, _) in enumerate(trading_days[5:]):
            if is_train and date >= split_date:
                continue
            if not is_train and date < split_date:
                continue
            
            if not is_train:
                valid_days_processed += 1
                logging.debug(f"Processing valid day {date}, count: {valid_days_processed}")
            
            current_data = {asset: market_data[asset].loc[:date] for asset in market_data}
            signals = strategy.generate_signals(current_data)
            trade_result = strategy.execute_trades(signals, current_data, 'train' if is_train else 'valid')
            new_trades = trade_result['trades']
            
            strategy.capital = trade_result['capital']
            self.capital = strategy.capital
            
            for trade in new_trades.get(model_type, []):
                pos_key = f"{model_type}_{trade['asset']}"
                if 'exit_price' in trade:
                    results['trades'][model_type].append(trade)
                    if pos_key in open_positions:
                        del open_positions[pos_key]
                elif trade['action'] in ['buy', 'sell']:
                    open_positions[pos_key] = trade
            
            # Корректный расчет portfolio_value
            portfolio_value = self.capital
            for pos_key, pos in open_positions.items():
                if pos['model'] == model_type:
                    current_price = current_data[pos['asset']]['Close'].iloc[-1]
                    if pos['action'] == 'buy':
                        # Для длинной позиции: текущая стоимость минус затраты на вход
                        position_value = pos['size'] * (current_price - pos['entry_price'])
                        portfolio_value += position_value
                    elif pos['action'] == 'sell':
                        # Для короткой позиции: прибыль или убыток от разницы цен
                        position_value = pos['size'] * (pos['entry_price'] - current_price)
                        portfolio_value += position_value
            
            # Убедимся, что portfolio_value не становится отрицательным без причины
            if portfolio_value < 0 and self.capital > 0:
                logging.warning(f"Portfolio value dropped below zero ({portfolio_value:.2f}) despite positive capital ({self.capital:.2f})")
            
            results['equity_curve'].append(portfolio_value)
            results['timestamps'].append(date)
            logging.info(f"Date={date}, PortfolioValue={portfolio_value:.4f}, Capital={self.capital:.4f}")
        
        metrics = self._calculate_metrics(results['equity_curve'], results['trades'][model_type], market_data)
        metrics['buy_and_hold_return'] = total_buy_and_hold_return
        metrics['trade_count'] = len(results['trades'][model_type])
        return {'metrics': metrics, 'results': results}
    
    def _calculate_metrics(self, equity_curve: List[float], trades: List[Dict], market_data: Dict[str, pd.DataFrame]) -> Dict:
        equity_series = pd.Series(equity_curve)
        returns = equity_series.pct_change().dropna()
    
        logging.debug(f"Equity series: {equity_series.tolist()}")
        logging.debug(f"Returns: {returns.tolist()}")
    
        risk_free_rate = 0.02 / 252
        mean_return = returns.mean() if not returns.empty else 0
        std_return = returns.std() if not returns.empty else 1e-6
        sharpe = ((mean_return - risk_free_rate) * np.sqrt(252)) / std_return if std_return != 0 else 0
    
        cumulative_returns = equity_series / equity_series.iloc[0]
        peak = cumulative_returns.cummax()
        drawdown = (cumulative_returns - peak) / peak
        max_drawdown = drawdown.min() * 100 if not drawdown.empty else 0
    
        total_return = ((equity_series.iloc[-1] / equity_series.iloc[0]) - 1) * 100 if len(equity_series) > 1 else 0
        total_return = min(total_return, 10000.0)
    
        profits = [t['profit_loss'] for t in trades if 'profit_loss' in t]
        wins = [p for p in profits if p > 0] or [0]
        losses = [abs(p) for p in profits if p < 0] or [0]  # Убытки как абсолютные значения для расчета
        total_wins = sum(wins)
        total_losses = sum(losses)
        profit_loss_ratio = total_wins / total_losses if total_losses > 0 else (total_wins if total_wins > 0 else 1.0)
    
        risk_reward_ratios = []
        for t in trades:
            if 'exit_price' in t and 'stop_loss' in t and 'take_profit' in t:
                risk = abs(t['entry_price'] - t['stop_loss'])
                reward = abs(t['take_profit'] - t['entry_price'])
                if risk > 0 and reward > 0:
                    risk_reward_ratios.append(reward / risk)
        risk_reward_ratio = np.mean(risk_reward_ratios) if risk_reward_ratios else 0
    
        win_rate = len(wins) / len(profits) * 100 if profits else 0
        calmar_ratio = (total_return / 100) / abs(max_drawdown / 100) if max_drawdown != 0 else 0
        calmar_ratio = min(calmar_ratio, 100.0)
    
        downside_returns = returns[returns < 0]
        downside_std = downside_returns.std() if not downside_returns.empty else 1e-6
        sortino = ((mean_return - risk_free_rate) * np.sqrt(252)) / downside_std if downside_std != 0 else 0
        sortino = min(sortino, 50.0)
    
        trade_count = len(profits)
        profitable_trades = len(wins)
        losing_trades = len([p for p in profits if p < 0])
    
        return {
            'sharpe_ratio': sharpe,
            'max_drawdown': max_drawdown,
            'total_return': total_return,
            'trade_count': trade_count,
            'profit_loss_ratio': profit_loss_ratio,
            'risk_reward_ratio': risk_reward_ratio,
            'win_rate': win_rate,
            'calmar_ratio': calmar_ratio,
            'sortino_ratio': sortino,
            'profitable_trades': profitable_trades,
            'losing_trades': losing_trades
        }

    def allocate_risk(self, positions):
        portfolio_risk = sum(pos['risk'] for pos in positions.values() if 'risk' in pos)
        if portfolio_risk > self.max_portfolio_risk:
            scale = self.max_portfolio_risk / portfolio_risk
            for pos in positions.values():
                pos['size'] *= scale
                pos['risk'] *= scale

    def _generate_trading_days(self, market_data):
        combined = pd.concat(market_data, axis=1, keys=market_data.keys(), names=['Asset', 'Feature']).ffill()
        business_days = pd.bdate_range(start=combined.index.min(), end=combined.index.max())
        for date in business_days:
            if date in combined.index:
                yield date, None
                
class EnhancedVisualizer:
    def create_dashboard(self, lstm_metrics: Dict, tft_metrics: Dict, lstm_results: Dict, tft_results: Dict, 
                        lstm_valid_results: Dict, tft_valid_results: Dict, lstm_model, tft_model, market_data: Dict[str, pd.DataFrame],
                        lstm_train_opt_results: Dict, tft_train_opt_results: Dict, best_config_lstm: Dict, best_config_tft: Dict):
        total_rows = 7
        specs = [[{"type": "xy"}, {"type": "xy"}] for _ in range(3)] + \
                [[{"type": "table"}, {"type": "table"}] for _ in range(3)] + \
                [[{"type": "table"}, {"type": "table"}]]
        
        fig = make_subplots(
            rows=total_rows, cols=2,
            subplot_titles=[
                "LSTM Equity (Train)", "TFT Equity (Train)",
                "LSTM Optimized Equity (Train)", "TFT Optimized Equity (Train)",
                "LSTM Equity (Valid with Opt)", "TFT Equity (Valid with Opt)",
                "LSTM Trades (Train)", "TFT Trades (Train)",
                "LSTM Optimized Trades (Train)", "TFT Optimized Trades (Train)",
                "LSTM Trades (Valid with Opt)", "TFT Trades (Valid with Opt)",
                "LSTM Optimal Parameters", "TFT Optimal Parameters"
            ],
            specs=specs,
            row_heights=[800] * total_rows,
            vertical_spacing=0.03
        )
        
        def add_equity_curve(results, model_name, row, col, period='train'):
            equity_timestamps = results['timestamps']
            equity_values = results['equity_curve']
            split_date = pd.Timestamp('2023-10-15')
            
            if not equity_timestamps or not equity_values:
                logging.warning(f"No equity data for {model_name} ({period})")
                equity_timestamps = [split_date if period == 'valid' else pd.Timestamp('2019-09-05')]
                equity_values = [INITIAL_CAPITAL]
            else:
                if period == 'valid':
                    mask = [t >= split_date for t in equity_timestamps]
                else:
                    mask = [t < split_date for t in equity_timestamps]
                filtered_timestamps = [t for t, m in zip(equity_timestamps, mask) if m]
                filtered_values = [v for v, m in zip(equity_values, mask) if m]
                
                if not filtered_timestamps:
                    logging.warning(f"No filtered equity data for {model_name} ({period})")
                    filtered_timestamps = [split_date if period == 'valid' else pd.Timestamp('2019-09-05')]
                    filtered_values = [INITIAL_CAPITAL]
                else:
                    start_date = split_date if period == 'valid' else pd.Timestamp('2019-09-05')
                    if filtered_timestamps[0] != start_date or abs(filtered_values[0] - INITIAL_CAPITAL) > 1e-6:
                        filtered_timestamps.insert(0, start_date)
                        filtered_values.insert(0, INITIAL_CAPITAL)
                    equity_timestamps = filtered_timestamps
                    equity_values = filtered_values
            
            fig.add_trace(go.Scatter(
                x=equity_timestamps, 
                y=equity_values, 
                name=f'{model_name} Equity ({period.capitalize()})', 
                mode='lines',
                line=dict(width=2)
            ), row=row, col=col)
        
        def add_trades_table(trades, row, col, model_name, period='train'):
            trade_rows = []
            split_date = pd.Timestamp('2023-10-15')
            
            for t in trades:
                trade_timestamp = pd.Timestamp(t['entry_timestamp'])
                is_valid_trade = (
                    (period == 'train' and trade_timestamp < split_date) or
                    (period == 'valid' and trade_timestamp >= split_date)
                )
                
                if is_valid_trade:
                    profit_loss_value = t.get('profit_loss', 0) if 'exit_price' in t else 0
                    profit_loss = f"{profit_loss_value:+.4f}"
                    display_timestamp = pd.Timestamp(t['exit_timestamp']) if 'exit_timestamp' in t else pd.Timestamp(t['entry_timestamp'])
                    
                    trade_rows.append([
                        t['asset'],
                        t['action'],
                        f"{t['size']:.4f}",
                        f"{t['entry_price']:.4f}",
                        f"{t.get('exit_price', 'N/A'):.4f}" if 'exit_price' in t else 'N/A',
                        t.get('close_reason', 'Open'),
                        profit_loss if 'exit_price' in t else 'N/A',
                        display_timestamp.strftime('%Y-%m-%d') if pd.notnull(display_timestamp) else 'N/A'
                    ])
            
            if not trade_rows:
                logging.info(f"No trades found for {model_name} ({period}). Adding placeholder.")
                default_date = '2023-10-15' if period == 'valid' else '2019-09-05'
                trade_rows.append(['N/A', 'hold', '0.0000', '0.0000', '0.0000', 'N/A', '+0.0000', default_date])
            
            headers = ['Asset', 'Action', 'Size', 'Entry Price', 'Exit Price', 'Reason', 'Profit/Loss', 'Date']
            fig.add_trace(go.Table(
                header=dict(values=headers),
                cells=dict(values=[list(col) for col in zip(*trade_rows)]),
                columnwidth=[550, 500, 600, 650, 650, 650, 650, 700]
            ), row=row, col=col)
        
        def add_optimal_parameters(row, col, model_name, optimal_params):
            params_rows = [[k, str(v)] for k, v in optimal_params.items()]
            headers = ['Parameter', 'Value']
            fig.add_trace(go.Table(
                header=dict(values=headers),
                cells=dict(values=[list(col) for col in zip(*params_rows)]),
                columnwidth=[300, 250]
            ), row=row, col=col)
        
        add_equity_curve(lstm_valid_results, 'LSTM', 3, 1, 'valid')
        add_equity_curve(tft_valid_results, 'TFT', 3, 2, 'valid')
        add_trades_table(lstm_valid_results['trades']['lstm'], 6, 1, 'LSTM', 'valid')
        add_trades_table(tft_valid_results['trades']['tft'], 6, 2, 'TFT', 'valid')
        add_equity_curve(lstm_results, 'LSTM', 1, 1, 'train')
        add_equity_curve(tft_results, 'TFT', 1, 2, 'train')
        add_equity_curve(lstm_train_opt_results, 'LSTM Optimized', 2, 1, 'train')
        add_equity_curve(tft_train_opt_results, 'TFT Optimized', 2, 2, 'train')
        add_trades_table(lstm_results['trades']['lstm'], 4, 1, 'LSTM', 'train')
        add_trades_table(tft_results['trades']['tft'], 4, 2, 'TFT', 'train')
        add_trades_table(lstm_train_opt_results['trades']['lstm'], 5, 1, 'LSTM Optimized', 'train')
        add_trades_table(tft_train_opt_results['trades']['tft'], 5, 2, 'TFT Optimized', 'train')
        add_optimal_parameters(7, 1, 'LSTM', best_config_lstm)
        add_optimal_parameters(7, 2, 'TFT', best_config_tft)
        
        fig.update_layout(
            height=2600,
            width=1800,
            title_text="Trading Strategy Dashboard",
            title_x=0.5,
            showlegend=True
        )
        
        if is_jupyter():
            display(fig)
        else:
            fig.write_image("trading_dashboard_full.png")
        logging.info("Full dashboard displayed")
    
def optimize_strategy_lstm(trial, datamodule, market_data, granger_results, corr_data, clusters, class_weights):
    config = {
        'confidence_threshold': trial.suggest_float('confidence', 0.3, 0.9),
        'initial_capital': INITIAL_CAPITAL,  # Используем глобальный INITIAL_CAPITAL
        'assets': datamodule.all_assets,
        'stop_loss_mult': trial.suggest_float('stop_loss_mult', 2.0, 4.0),
        'take_profit_mult': trial.suggest_float('take_profit_mult', 4.0, 9.0),
        'size_percent': trial.suggest_float('size_percent', 0.05, 0.5)
    }
    lstm_model = EnhancedLSTMModel(input_size=14, class_weights=torch.tensor(class_weights, dtype=torch.float))
    tft_model = TemporalFusionTransformerModel(datamodule.train_dataset, class_weights=torch.tensor(class_weights, dtype=torch.float))
    trainer = pl.Trainer(max_epochs=5, accelerator='auto', logger=False, enable_progress_bar=False)
    trainer.fit(lstm_model, datamodule)
    trainer.fit(tft_model, datamodule)
    strategy = EnhancedTradingStrategy(lstm_model, tft_model, config, datamodule.asset_encoder, granger_results, corr_data, clusters)
    evaluator = EnhancedStrategyEvaluator(config)
    results = evaluator.evaluate(strategy, market_data, datamodule.train_dataset, max_days=1500, model_type='lstm')
    sharpe = results['metrics'].get('sharpe_ratio', 0.0)
    logging.info(f"LSTM Trial {trial.number} - Sharpe Ratio: {sharpe}")
    return sharpe

def optimize_strategy_tft(trial, datamodule, market_data, granger_results, corr_data, clusters, class_weights):
    config = {
        'confidence_threshold': trial.suggest_float('confidence', 0.3, 0.9),
        'initial_capital': INITIAL_CAPITAL,  # Используем глобальный INITIAL_CAPITAL
        'assets': datamodule.all_assets,
        'stop_loss_mult': trial.suggest_float('stop_loss_mult', 2.0, 4.0),
        'take_profit_mult': trial.suggest_float('take_profit_mult', 4.0, 9.0),
        'size_percent': trial.suggest_float('size_percent', 0.05, 0.5)
    }
    lstm_model = EnhancedLSTMModel(input_size=14, class_weights=torch.tensor(class_weights, dtype=torch.float))
    tft_model = TemporalFusionTransformerModel(datamodule.train_dataset, class_weights=torch.tensor(class_weights, dtype=torch.float))
    trainer = pl.Trainer(max_epochs=5, accelerator='auto', logger=False, enable_progress_bar=False)
    trainer.fit(lstm_model, datamodule)
    trainer.fit(tft_model, datamodule)
    strategy = EnhancedTradingStrategy(lstm_model, tft_model, config, datamodule.asset_encoder, granger_results, corr_data, clusters)
    evaluator = EnhancedStrategyEvaluator(config)
    results = evaluator.evaluate(strategy, market_data, datamodule.train_dataset, max_days=500, model_type='tft')
    sharpe = results['metrics'].get('sharpe_ratio', 0.0)
    logging.info(f"TFT Trial {trial.number} - Sharpe Ratio: {sharpe}")
    return sharpe

def main():
    try:
        loader = EnhancedDataLoader(
            stocks=['MSTR', 'MARA', 'GBTC', 'BITF'],
            cryptos=['BTC-USD', 'ETH-USD']
        )
        raw_data = loader.load_data()
        preprocessor = AdvancedPreprocessor()
        processed_data = preprocessor.process(raw_data['raw'], all_assets=loader.all_assets)

        # Отображение обработанных данных
        visualizer = EnhancedVisualizer()
        
        analyzer = EnhancedMarketAnalyzer()
        granger_data = pd.concat({k: v['Close'] for k, v in processed_data.items()}, axis=1)
        granger_results = analyzer.granger_analysis(granger_data)
        corr_pearson, corr_spearman = analyzer.correlation_analysis(granger_data)
        clusters = analyzer.cluster_analysis(granger_data)
        combined_data = pd.concat([df.assign(asset=k) for k, df in processed_data.items()])
        datamodule = TimeSeriesDataModule(combined_data, all_assets=loader.all_assets)
        datamodule.setup()
        class_weights = compute_class_weight('balanced', classes=np.array([0, 1, 2]), y=datamodule.data['target'].values)

        lstm_model = EnhancedLSTMModel(input_size=14, class_weights=torch.tensor(class_weights, dtype=torch.float))
        tft_model = TemporalFusionTransformerModel(datamodule.train_dataset, class_weights=torch.tensor(class_weights, dtype=torch.float))
        trainer = pl.Trainer(
            max_epochs=5,
            callbacks=[EarlyStopping(monitor='val_loss', patience=15, min_delta=0.0001)],
            accelerator='auto',
            precision=16 if torch.cuda.is_available() else 32,
            logger=True,
            check_val_every_n_epoch=2
        )
        trainer.fit(lstm_model, datamodule)
        trainer.fit(tft_model, datamodule)

        # Базовая конфигурация
        config = {
            'initial_capital': INITIAL_CAPITAL,
            'confidence_threshold': 0.3,
            'stop_loss_mult': 2.0,
            'take_profit_mult': 4.0,
            'size_percent': 0.05,
            'assets': loader.all_assets
        }
        
        # Создание стратегии и оценщика до использования
        strategy = EnhancedTradingStrategy(lstm_model, tft_model, config, datamodule.asset_encoder, granger_results, (corr_pearson, corr_spearman), clusters)
        evaluator = EnhancedStrategyEvaluator(config)

        # Бэктестирование на train-наборе (неоптимизированное)
        logging.info("Starting LSTM train backtest")
        lstm_train_results = evaluator.evaluate(strategy, processed_data, datamodule.train_dataset, model_type='lstm', max_days=1400)
        logging.info(f"LSTM Train Capital after backtest: {evaluator.capital}")
    
        logging.info("Starting TFT train backtest")
        tft_train_results = evaluator.evaluate(strategy, processed_data, datamodule.train_dataset, model_type='tft', max_days=500)
        logging.info(f"TFT Train Capital after backtest: {evaluator.capital}")

        # Оптимизация параметров на train-наборе
        logging.info("Starting LSTM optimization")
        study_lstm = optuna.create_study(direction='maximize')
        study_lstm.optimize(lambda trial: optimize_strategy_lstm(trial, datamodule, processed_data, granger_results, (corr_pearson, corr_spearman), clusters, class_weights), n_trials=3)
        best_config_lstm = study_lstm.best_params
        config_lstm = {
            'initial_capital': INITIAL_CAPITAL,
            'confidence_threshold': best_config_lstm['confidence'],
            'stop_loss_mult': best_config_lstm['stop_loss_mult'],
            'take_profit_mult': best_config_lstm['take_profit_mult'],
            'size_percent': best_config_lstm['size_percent'],
            'assets': loader.all_assets
        }

        logging.info("Starting TFT optimization")
        study_tft = optuna.create_study(direction='maximize')
        study_tft.optimize(lambda trial: optimize_strategy_tft(trial, datamodule, processed_data, granger_results, (corr_pearson, corr_spearman), clusters, class_weights), n_trials=3)
        best_config_tft = study_tft.best_params
        config_tft = {
            'initial_capital': INITIAL_CAPITAL,
            'confidence_threshold': best_config_tft['confidence'],
            'stop_loss_mult': best_config_tft['stop_loss_mult'],
            'take_profit_mult': best_config_tft['take_profit_mult'],
            'size_percent': best_config_tft['size_percent'],
            'assets': loader.all_assets
        }

        # Бэктестирование на train-наборе с оптимальными параметрами
        strategy_lstm_opt = EnhancedTradingStrategy(lstm_model, tft_model, config_lstm, datamodule.asset_encoder, granger_results, (corr_pearson, corr_spearman), clusters)
        strategy_tft_opt = EnhancedTradingStrategy(lstm_model, tft_model, config_tft, datamodule.asset_encoder, granger_results, (corr_pearson, corr_spearman), clusters)
        evaluator_lstm_opt = EnhancedStrategyEvaluator(config_lstm)
        evaluator_tft_opt = EnhancedStrategyEvaluator(config_tft)

        logging.info("Starting LSTM optimized train backtest")
        lstm_train_opt_results = evaluator_lstm_opt.evaluate(strategy_lstm_opt, processed_data, datamodule.train_dataset, model_type='lstm', max_days=1400)
        logging.info("Starting TFT optimized train backtest")
        tft_train_opt_results = evaluator_tft_opt.evaluate(strategy_tft_opt, processed_data, datamodule.train_dataset, model_type='tft', max_days=500)

        # Бэктестирование на valid-наборе с оптимальными параметрами
        logging.info("Starting LSTM valid backtest")
        lstm_valid_results = evaluator_lstm_opt.evaluate(strategy_lstm_opt, processed_data, datamodule.val_dataset, is_train=False, model_type='lstm', max_days=1400)
        logging.info("Starting TFT valid backtest")
        tft_valid_results = evaluator_tft_opt.evaluate(strategy_tft_opt, processed_data, datamodule.val_dataset, is_train=False, model_type='tft', max_days=500)
        
        # Логирование результатов
        logging.info("LSTM Train Metrics: %s", lstm_train_results['metrics'])
        logging.info("TFT Train Metrics: %s", tft_train_results['metrics'])
        logging.info("LSTM Train Optimized Metrics: %s", lstm_train_opt_results['metrics'])
        logging.info("TFT Train Optimized Metrics: %s", tft_train_opt_results['metrics'])
        logging.info("LSTM Valid Optimized Metrics: %s", lstm_valid_results['metrics'])
        logging.info("TFT Valid Optimized Metrics: %s", tft_valid_results['metrics'])

        # Метрики модели
        def get_model_metrics(trainer, model_type, dataset):
            trainer.validate(model=lstm_model if model_type == 'lstm' else tft_model, dataloaders=dataset)
            return {
                'Accuracy': trainer.logged_metrics.get('val_accuracy', 0.0),
                'Precision': trainer.logged_metrics.get('val_precision', 0.0),
                'Recall': trainer.logged_metrics.get('val_recall', 0.0),
                'F1': trainer.logged_metrics.get('val_f1', 0.0),
                'ROC-AUC': trainer.logged_metrics.get('val_roc_auc', 0.0)
            }

        metrics_table = {
            'Metrics': [
                'buy_and_hold_return',
                'trade_count',
                'profitable_trades',
                'losing_trades',
                'sharpe_ratio',
                'max_drawdown',
                'total_return',
                'profit_loss_ratio',
                'risk_reward_ratio',
                'win_rate',
                'calmar_ratio',
                'sortino_ratio',
                'Accuracy',
                'Precision',
                'Recall',
                'F1',
                'ROC-AUC'
            ]
        }

        columns = ['LSTM Train', 'TFT Train', 'LSTM Train Opt', 'TFT Train Opt', 'LSTM Valid with Opt', 'TFT Valid with Opt']
        for col in columns:
            if col == 'LSTM Train':
                metrics = lstm_train_results['metrics']
                model_metrics = get_model_metrics(trainer, 'lstm', datamodule.train_dataloader())
            elif col == 'TFT Train':
                metrics = tft_train_results['metrics']
                model_metrics = get_model_metrics(trainer, 'tft', datamodule.train_dataloader())
            elif col == 'LSTM Train Opt':
                metrics = lstm_train_opt_results['metrics']
                model_metrics = get_model_metrics(trainer, 'lstm', datamodule.train_dataloader())
            elif col == 'TFT Train Opt':
                metrics = tft_train_opt_results['metrics']
                model_metrics = get_model_metrics(trainer, 'tft', datamodule.train_dataloader())
            elif col == 'LSTM Valid with Opt':
                metrics = lstm_valid_results['metrics']
                model_metrics = get_model_metrics(trainer, 'lstm', datamodule.val_dataloader())
            elif col == 'TFT Valid with Opt':
                metrics = tft_valid_results['metrics']
                model_metrics = get_model_metrics(trainer, 'tft', datamodule.val_dataloader())

            values = [
                f"{metrics['buy_and_hold_return']:.4f}",
                f"{int(metrics['trade_count'])}",
                f"{metrics.get('profitable_trades', 0)}",
                f"{metrics.get('losing_trades', 0)}",
                f"{metrics['sharpe_ratio']:.4f}",
                f"{metrics['max_drawdown']:.4f}",
                f"{metrics['total_return']:.4f}",
                f"{metrics['profit_loss_ratio']:.4f}",
                f"{metrics['risk_reward_ratio']:.4f}",
                f"{metrics['win_rate']:.4f}",
                f"{metrics['calmar_ratio']:.4f}",
                f"{metrics['sortino_ratio']:.4f}",
                f"{model_metrics['Accuracy']:.4f}",
                f"{model_metrics['Precision']:.4f}",
                f"{model_metrics['Recall']:.4f}",
                f"{model_metrics['F1']:.4f}",
                f"{model_metrics['ROC-AUC']:.4f}"
            ]
            metrics_table[col] = values
        
        metrics_df = pd.DataFrame(metrics_table).set_index('Metrics')
        if is_jupyter():
            display(HTML("Performance and Model Metrics"))
            display(metrics_df)
        else:
            print(f"\n\033[1;34m Performance and Model Metrics \033[0m")
            print(metrics_df.to_string())

        # Создание дашборда
        visualizer.create_dashboard(
            lstm_train_results['metrics'], tft_train_results['metrics'],
            lstm_train_results['results'], tft_train_results['results'],
            lstm_valid_results['results'], tft_valid_results['results'],
            lstm_model, tft_model, 
            processed_data,
            lstm_train_opt_results['results'], tft_train_opt_results['results'],
            best_config_lstm, best_config_tft
        )

        logging.info("Execution completed successfully")
        print("Execution completed. Check output in notebook or console and 'training.log'.")
    except Exception as e:
        logging.error(f"Main failed: {str(e)}")
        print(f"Error: {str(e)}. Check 'training.log'.")

if __name__ == "__main__":
    main()
```
